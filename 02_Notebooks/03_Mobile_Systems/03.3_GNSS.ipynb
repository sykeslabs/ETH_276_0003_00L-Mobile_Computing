### 3.1 Introduction

For centuries, determining one’s exact time and position was one of the greatest challenges for human civilizations. Early navigators relied on natural references such as the Sun, the Moon, and the stars to guide their journeys across land and sea. These celestial bodies provided approximate orientation and a sense of time, but the accuracy was limited and highly dependent on weather, visibility, and skill in observation. The development of mechanical clocks and sextants in the Age of Exploration greatly improved navigation, especially at sea, by allowing sailors to measure angles between celestial bodies and the horizon. However, precise determination of longitude remained a problem well into the 18th century, only solved with the invention of highly accurate marine chronometers. Today, this age-old quest for accurate time and position has been revolutionized by satellite navigation systems. By measuring the time it takes for signals to travel from satellites in orbit to receivers on Earth, humans can now determine their location with meter- or even centimeter-level precision - anywhere on the planet, at any time, and in almost any weather.

In this notebook, we will explore Global Navigation Satellite System (GNSS) technology with a focus on the Global Positioning System (GPS). We will examine how position and time are determined using trilateration, the radio frequencies employed by the system, and the way signals are transmitted, received, and interpreted, including the information they carry.

### 3.2 Global Navigation Satellite System - GNSS [[01](#refAAA)]

#### 3.2.1 GNSS and GPS

The term GNSS (Global Navigation Satellite System) refers to the overall concept of satellite-based positioning, navigation, and timing (PNT) systems that provide global coverage. A GNSS is not a single system but rather an umbrella term for all constellations of satellites that transmit signals allowing users on Earth to determine their position and precise time. GPS (Global Positioning System), by contrast, is one specific implementation of a GNSS. Developed and operated by the United States Department of Defense, GPS was the first fully operational satellite navigation system and remains the most widely used today. Because of its early dominance, the term “GPS” is often used colloquially to mean “satellite navigation” in general - but strictly speaking, it is just one component of the broader GNSS landscape. A GNSS receiver is typically capable of using signals from multiple constellations simultaneously. For example, modern smartphones and surveying equipment often combine signals from GPS and Galileo, and sometimes also GLONASS or BeiDou. This multi-constellation capability increases accuracy, improves reliability, and enhances performance in challenging environments such as urban canyons or mountainous terrain, where visibility of satellites may be limited.


#### 3.2.2 Major GNSS systems

The main global GNSS constellations in operation today are:

* GPS (United States): The oldest and most widely adopted system, operational since the 1990s.
* Galileo (European Union): Europe’s independent GNSS, designed for high accuracy and civilian control.
* GLONASS (Russia): Developed by the Soviet Union and later modernized by Russia, offering full global coverage.
* BeiDou (China): China’s navigation system, expanded to global coverage in 2020, with unique regional augmentation services.

Together, these systems form the backbone of today’s GNSS services, enabling precise positioning and timing worldwide for applications ranging from navigation apps in smartphones to aviation, maritime operations, agriculture, and scientific research.

#### 3.2.3 Orbital speeds, heights, periods

The orbits of various satellite systems are shown in the image below.

A geostationary orbit is located about 35,786 km above Earth’s equator. A satellite here takes exactly 24 hours to complete one orbit, which matches Earth’s rotation. From the ground, the satellite appears fixed in the sky, always above the same point. This makes it ideal for television broadcasting, weather monitoring, and communications, since ground antennas do not need to move. One GEO satellite can cover roughly a third of Earth’s surface, so three satellites provide near-global coverage. The trade-off is latency: signals take about a quarter of a second to travel one way, which can be noticeable in real-time communication.

A Low Earth Orbit lies much closer to Earth, between 300 and 2,000 km altitude. Satellites here orbit very quickly, typically in about 90 minutes. Because they move fast relative to Earth’s surface, a single satellite cannot provide continuous coverage - large constellations are required. Modern systems like Starlink or OneWeb rely on hundreds or thousands of LEO satellites to ensure constant internet connectivity. The advantage is very low latency (under 30 milliseconds) and strong signal power because of the short distance. The downside is the cost and complexity of maintaining such a large constellation, along with the growing risk of space debris.

Indicated in the image is also the inner Van Allen belt. Surrounding Earth are two regions filled with charged particles trapped by the planet’s magnetic field. These are the two Van Allen radiation belts. The inner belt extends from roughly 600 to 12,000 km altitude, while the outer belt ranges from about 13,000 to 60,000 km. Satellites in these zones face a harsh radiation environment that can damage electronics and shorten lifespans, unless they are shielded or radiation-hardened. Satellites in LEO orbit safely below the belts, and satellites in GEO orbit sit well above them. However, spacecraft in medium Earth orbit, such as GPS satellites at around 20,200 km, must regularly operate within the belts and therefore need special protection.

#### 3.2.4 Satellite lifetimes

Most satellites have a limited lifetime of 5 to 20 years, depending on their orbit, propulsion, and design. In Low Earth Orbit (LEO), satellites gradually lose altitude due to atmospheric drag, so their natural lifetime can range from a few months (below 400 km) up to about 10 years (500–600 km). With propulsion, operators can extend service life or control reentry. In medium Earth orbit (MEO) and geostationary orbit (GEO), where drag is negligible, satellites typically last 10–20 years, limited mainly by onboard fuel, radiation damage, and aging electronics. At end of life, LEO satellites are deorbited to burn up in the atmosphere, while GEO satellites are moved to a “graveyard orbit.”

The popular Starlink satellites in Low Earth Orbit (~550 km) have an operational lifetime of about 5 years. They use ion thrusters to maintain orbit and are designed to deorbit at the end of life, burning up in Earth’s atmosphere within a year. Without propulsion, they would naturally fall back in about 5–7 years due to [atmospheric drag](https://www.sws.bom.gov.au/Educational/1/3/8). This shorter lifespan (compared to 15–20 years for geostationary satellites) is intentional, as Starlink satellites are built for rapid replacement in a large constellation.
>Starlink satellites are not designed for navigation like GPS or Galileo, since they orbit much lower (~550 km vs. ~20,000 km), lack onboard atomic clocks, and transmit broadband signals rather than precise ranging codes. However, researchers have shown that Starlink signals can be used experimentally for positioning by measuring timing or Doppler shifts, and the system could serve as a useful GNSS complement or backup, especially for providing correction data or resilience against jamming. ### 3.3 Trilateration

#### 3.3.1 Estimating distance

GNSS satellites transmit electromagnetic signals that travel through space at the speed of light, $c = 300{,}000 \ \text{km/s}$. By measuring how long a signal takes to travel from the satellite to the receiver, we can estimate the distance as
$$
\text{Distance} = c \times (\text{Receive Time} - \text{Transmit Time}).
$$

To make this calculation possible, each satellite continuously broadcasts both its precise orbital position and a timestamp marking the exact moment of transmission. When the receiver picks up the signal, it compares the transmit timestamp with its own receive time. Knowing the distance to a single satellite means the receiver must be located somewhere on a sphere centered at that satellite’s position. To determine an exact location, signals from multiple satellites are required.

The precision of GNSS positioning is strongly influenced by signal bandwidth. A wider bandwidth produces sharper correlation peaks in the receiver, which improves the accuracy of measuring the signal’s travel time. The original GPS L1 C/A code has a bandwidth of about 2 MHz, yielding meter-level precision. In contrast, modern signals such as GPS L5 or Galileo E5 use chip rates around 10 MHz and bandwidths near 20 MHz, allowing sub-meter precision under good conditions (wide-band GNSS).

Even higher accuracy can be achieved with correction techniques. RTK (Real-Time Kinematic) uses carrier-phase measurements and reference stations to provide centimeter-level positioning in real time. PPP (Precise Point Positioning) applies precise satellite orbit and clock corrections, usually delivered via data services, to reach decimeter- to centimeter-level accuracy without needing a local base station. Both methods rely on the fine timing resolution made possible by wide-band GNSS signals.


#### 3.3.2 Multi-lateration

To determine a three-dimensional position (latitude, longitude, and altitude), a GNSS receiver must also solve for the offset of its internal clock relative to GNSS time. This requires four unknowns: three for position and one for clock bias. Each satellite measurement defines a sphere centered at the satellite’s position with a radius equal to the measured distance (pseudorange). The intersection of three such spheres would be enough if the receiver’s clock were perfectly synchronized, but since it is not, a fourth measurement is needed. In practice, signals from at least four satellites are required to compute an accurate 3D position and correct the receiver clock.

>The distance equation assumes that the satellite and the receiver are perfectly synchronized—that both use the same clock reference. In practice, this is not the case. Even a tiny time offset leads to large errors in position because signals travel at the speed of light ($c$). A clock offset of just 1 ms (one thousandth of a second) would result in a position error of about 300 km. GNSS satellites carry highly stable atomic clocks that remain accurate to nanoseconds, but receivers such as smartphones rely on inexpensive quartz oscillators, which can drift by microseconds or more. This clock drift introduces an unknown time bias between the receiver’s clock and true GNSS time. To solve for this bias, the receiver estimates its own clock offset along with its position. That adds a fourth unknown, so at least four satellites are required: three to determine the receiver’s position and one more to correct its internal clock. In practice, additional satellites provide extra measurements, which reduce errors, improve precision, and make the final position estimate more robust against noise and signal disturbances.


Modern GNSS receivers, however, almost always use more than four satellites. Depending on the location and visibility, a receiver typically tracks between 7 and 12 satellites from one constellation, and even more when combining multiple constellations such as GPS, Galileo, BeiDou, and GLONASS. Using additional satellites improves accuracy, reduces sensitivity to errors, and increases robustness in challenging environments.

### 3.4 Transmitting GPS signals

#### 3.4.1 GPS orbit and frequencies

GPS satellites orbit the Earth at an altitude of about 20,200 km, completing two orbits per day. Their motion is carefully synchronized so that at least four satellites are visible from anywhere on Earth at all times. Each satellite continuously transmits radio signals toward Earth. Only the satellites transmit. Receivers on the ground are passive; they only listen to these signals. GPS uses several carrier frequencies on different radio bands:

| Band | Frequency (MHz) | Use | Notes |
|------|-----------------|------|-------|
| **L1** | 1575.42 | Civil | Used by all standard GPS receivers |
| **L2** | 1227.60 | Military (and dual-frequency civil) | Helps reduce atmospheric errors |
| **L5** | 1176.45 | Civil (aviation, safety-of-life) | Higher accuracy, modern systems |

These signals travel over 20,000 km, so they become very weak by the time they reach the ground, around –160 dBW, much weaker than background noise.  However, GPS receivers can still detect them thanks to their unique coded structure and correlation processing, which allows the weak signals to stand out.

#### 3.4.2 Navigation message content

Each GPS signal includes a navigation message that carries important information about the satellite and the system. This message is transmitted at a very low data rate (50 b/s) and repeats continuously. It contains the **satellite status and health**, **precise time information**, **orbital parameters (ephemeris)** describing the exact path of the satellite, and **approximate orbit data (almanac)**, which is coarse information for all satellites. This message allows the receiver to compute where each satellite was when its signal was transmitted, which is essential for accurate positioning.

The almanac tells a receiver roughly where and when each satellite will be visible, but not with high precision. Each satellite broadcasts the full almanac for the whole constellation. A receiver uses it for initial acquisition (finding satellites quickly), but for accurate positioning it needs the ephemeris data (precise orbit info for a single satellite).

| **Aspect**       | **Almanac** (approximate orbit data)                                       | **Ephemeris** (precise orbit data)             |
| ---------------- | -------------------------------------------------------------------------- | ---------------------------------------------- |
| **Scope**        | Info for *all satellites* in the constellation                             | Info for *one satellite* only                  |
| **Accuracy**     | Coarse (km-level) orbit + clock info                                       | Precise (meter-level) orbit + clock info       |
| **Size**         | Small dataset (~1 KB)                                                      | Larger dataset (~1–2 KB per satellite)         |
| **Validity**     | Days to weeks (stable over time)                                           | A few hours (must be refreshed often)          |
| **Transmission** | Sent by every satellite, for all satellites                                | Sent by each satellite, only for itself        |
| **Use**          | Helps receiver know *which satellites* to search for, and where in the sky | Needed for *accurate positioning* calculations |
| **Acquisition**  | Takes ~12.5 minutes to receive full set                                    | Takes ~30 seconds per satellite (good signal)  |

#### 3.4.3 PRN Codes and signal modulation

Each GPS satellite uses a unique Pseudo-Random Noise (PRN) code to identify its signal. The main code used for civil applications is the C/A (Coarse/Acquisition) code, transmitted on the L1 frequency. The C/A code is a sequence of 1,023 binary values (+1 and –1) that repeats every 1 millisecond. It appears random, but is actually deterministic and publicly known - meaning every receiver can generate it locally. Each satellite has a different PRN code, allowing multiple satellites to share the same frequency (a principle called Code Division Multiple Access, CDMA). The GPS signal is modulated by combining a carrier wave (e.g., L1 at 1575.42 MHz), the C/A code (which spreads the signal spectrum), and the navigation message (slow data that carries timing and orbital info). At the receiver, the incoming signal is cross-correlated with a locally generated copy of the same PRN code. When the two codes align, a correlation peak appears, revealing the signal delay, which corresponds to the distance to that satellite.

#### 3.4.4 Demonstration: generating and correlating a PRN code
Below we simulate, in a simplified way, how a GPS receiver identifies a satellite signal:
1. Generate a pseudo-random (PRN) code sequence.  
2. Modulate it onto a carrier wave.  
3. Delay it to mimic transmission through space.  
4. Correlate the received signal with a local copy to recover the delay (i.e., distance).


##### 3.4.4.1 GPS satellite transmitter

```python
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# Parameters for visualization
# ----------------------------
num_symbols = 6        # show about 5–6 bits
samples_per_symbol = 1000
samples_per_chip = 100
fc = 100               # carrier frequency [Hz]
fs = 10000             # sampling frequency [Hz]
T_total = num_symbols * samples_per_symbol / fs
t = np.arange(0, T_total, 1/fs)

# Derived rates
Rs = fs / samples_per_symbol   # symbol rate
Rc = fs / samples_per_chip     # chip rate

# (optional) reproducibility
np.random.seed(0)

# ----------------------------
# Colors / styles
# ----------------------------
line_color = (1, 0, 1)           # magenta
axes_facecolor = (0.2, 0.2, 0.3) # dark bluish background

# ----------------------------
# 1) Transmitter "step" message (+1 / -1)
# ----------------------------
msg_symbols = np.array([1 if i % 2 == 0 else -1 for i in range(num_symbols)])
message = np.repeat(msg_symbols, samples_per_symbol)
message = message[:len(t)]

# ----------------------------
# 2) PRN code (+1 / -1)
# ----------------------------
num_chips = int(T_total * Rc)
prn_chips = np.random.choice([1, -1], size=num_chips)
prn = np.repeat(prn_chips, samples_per_chip)
prn = prn[:len(t)]

# ----------------------------
# 3) Spread signal (message × PRN)
# ----------------------------
spread = message * prn

# ----------------------------
# 4) Carrier
# ----------------------------
carrier = np.cos(2*np.pi*fc*t)

# ----------------------------
# 5) Modulated signal: (spread) × (carrier)
# ----------------------------
tx = spread * carrier

# ----------------------------
# Plot — aligned
# ----------------------------
fig, axes = plt.subplots(
    5, 1, figsize=(11, 8), sharex=True,
    gridspec_kw={"hspace": 0.12}
)

def draw_symbol_edges(ax):
    for k in range(num_symbols + 1):
        ax.axvline(k * samples_per_symbol / fs, linestyle='--', color='gray', alpha=0.3)
    ax.grid(True, which='both', axis='y', alpha=0.3)
    ax.set_facecolor(axes_facecolor)

# 1) Message
axes[0].step(t, message, where='post', linewidth=1.2, color=line_color)
axes[0].set_ylabel("Navigation Message\n(+1/−1)")
draw_symbol_edges(axes[0])

# 2) PRN
axes[1].step(t, prn, where='post', linewidth=1.2, color=line_color)
axes[1].set_ylabel("PRN\n(+1/−1)")
draw_symbol_edges(axes[1])

# 3) Spread (message × PRN)
axes[2].step(t, spread, where='post', linewidth=1.2, color=line_color)
axes[2].set_ylabel("Spread\n(message×PRN)")
draw_symbol_edges(axes[2])

# 4) Carrier
axes[3].plot(t, carrier, linewidth=1.0, color=line_color)
axes[3].set_ylabel("Carrier\ncos(2πf_ct)")
draw_symbol_edges(axes[3])

# 5) RF signal (spread × carrier)
axes[4].plot(t, tx, linewidth=1.0, color=line_color)
axes[4].set_ylabel("RF: spread×carrier")
axes[4].set_xlabel("Time [s]")
draw_symbol_edges(axes[4])

#fig.suptitle("DSSS Transmitter Chain (6 slow bits, aligned view)", y=0.995)
plt.show()
```

##### 3.4.4.2 GPS receiver

The code cell below simulates a received RF signal with a PRN code-phase offset, carrier phase offset, and AWGN. The visualization wipes off the carrier with a local oscillator (cos(2πf_ct)). The code searches code phase over 0..samples_per_chip-1 to find the best alignment. Finally, the code despreads & integrates per symbol, then does a simple sign decision to recover the message and reports the bit error rate.

```python
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# Parameters for visualization
# ----------------------------
num_symbols = 6        # show about 5–6 bits
samples_per_symbol = 1000
samples_per_chip = 100
fc = 100               # carrier frequency [Hz]
fs = 10000             # sampling frequency [Hz]
T_total = num_symbols * samples_per_symbol / fs
t = np.arange(0, T_total, 1/fs)

# Derived rates
Rs = fs / samples_per_symbol   # symbol rate
Rc = fs / samples_per_chip     # chip rate

# (optional) reproducibility
np.random.seed(0)

# ----------------------------
# Colors / styles (as requested)
# ----------------------------
line_color = (1, 0, 1)           # common line color (magenta)
axes_facecolor = (0.2, 0.2, 0.3) # common facecolor for all subplots

# ----------------------------
# 1) Transmitter "step" message (+1 / -1)
# ----------------------------
msg_symbols = np.array([1 if i % 2 == 0 else -1 for i in range(num_symbols)])
message = np.repeat(msg_symbols, samples_per_symbol)
message = message[:len(t)]

# ----------------------------
# 2) PRN code (+1 / -1)
# ----------------------------
num_chips = int(T_total * Rc)
prn_chips = np.random.choice([1, -1], size=num_chips)
prn = np.repeat(prn_chips, samples_per_chip)
prn = prn[:len(t)]

# ----------------------------
# 3) Spread signal (message × PRN)
# ----------------------------
spread = message * prn

# ----------------------------
# 4) Carrier
# ----------------------------
carrier = np.cos(2*np.pi*fc*t)

# ----------------------------
# 5) Modulated signal: (spread) × (carrier)
# ----------------------------
tx = spread * carrier

# ==========================================================
# RECEIVER SIDE: "combined received signal" + correlator
# ==========================================================

# --- Channel impairments / combination settings ---
snr_db = 0.0           # AWGN SNR in dB (wrt tx power)
code_phase_offset = 12 # samples (0..samples_per_chip-1) to misalign PRN
carrier_phase = 0.4    # radians phase offset in the channel
freq_offset_hz = 0.0   # small Doppler (keep 0 for simplicity)

# Apply PRN code-phase offset at the transmitter path to simulate misalignment at the RX
prn_rx = np.roll(prn, code_phase_offset)
spread_rx = message * prn_rx

# Apply carrier with channel phase/freq offset
carrier_ch = np.cos(2*np.pi*(fc + freq_offset_hz)*t + carrier_phase)
tx_offset = spread_rx * carrier_ch

# Add AWGN
def add_awgn(x, snr_db):
    sig_power = np.mean(x**2)
    snr_lin = 10**(snr_db/10)
    noise_power = sig_power / snr_lin
    noise = np.sqrt(noise_power) * np.random.randn(*x.shape)
    return x + noise

rx = add_awgn(tx_offset, snr_db)

# --- Receiver: carrier wipeoff (assume known fc; unknown phase reduces amplitude) ---
lo = np.cos(2*np.pi*fc*t)  # if freq_offset_hz != 0 or phase != 0 -> residual loss
baseband = rx * lo         # coherent downconversion to (roughly) baseband

# --- Code-phase search (0..samples_per_chip-1) using short window ---
# We use a short window (e.g., 1 symbol) for the correlation metric
win_len = samples_per_symbol  # integrate over one symbol worth of samples

metrics = np.zeros(samples_per_chip)
for tau in range(samples_per_chip):
    prn_hyp = np.roll(prn, tau)
    # Despread hypothesis over window
    y = baseband[:win_len] * prn_hyp[:win_len]
    metrics[tau] = np.abs(np.sum(y))  # magnitude correlation metric

est_code_phase = int(np.argmax(metrics))

# --- Despread with estimated code phase over full duration ---
prn_aligned = np.roll(prn, est_code_phase)
despread = baseband * prn_aligned

# --- Symbol integrate-and-dump to recover message ---
despread_reshaped = despread[:num_symbols*samples_per_symbol].reshape(num_symbols, samples_per_symbol)
symbol_sums = np.sum(despread_reshaped, axis=1)
msg_hat = np.sign(symbol_sums)
# In case any zero (exactly), map to +1
msg_hat[msg_hat == 0] = 1

# --- BER (expect 0 with enough SNR and perfect carrier wipeoff) ---
ber = np.mean(msg_hat != msg_symbols)

# ----------------------------
# Plot — TX chain + RX correlator results
# ----------------------------
fig, axes = plt.subplots(
    7, 1, figsize=(12, 16), sharex=False,
    gridspec_kw={"hspace": 0.8}
)

def beautify(ax):
    ax.grid(True, which='both', axis='y', alpha=0.3)
    ax.set_facecolor(axes_facecolor)

# TX message
axes[0].step(t, message, where='post', linewidth=1.2, color=line_color)
axes[0].set_ylabel("Message\n(+1/−1)")
#axes[0].set_title("DSSS TX, Channel, and RX Correlator")
beautify(axes[0])

# TX PRN
axes[1].step(t, prn, where='post', linewidth=1.2, color=line_color)
axes[1].set_ylabel("PRN\n(+1/−1)")
beautify(axes[1])

# TX spread
axes[2].step(t, spread, where='post', linewidth=1.2, color=line_color)
axes[2].set_ylabel("Spread\n(m×PRN)")
beautify(axes[2])

# RX signal (after channel + noise)
axes[3].plot(t, rx, linewidth=0.9, color=line_color)
axes[3].set_ylabel("RX (RF)")
beautify(axes[3])

# Baseband after wipeoff
axes[4].plot(t, baseband, linewidth=0.9, color=line_color)
axes[4].set_ylabel("Baseband\n(wipeoff)")
beautify(axes[4])

# Code-phase correlation metric
axes[5].plot(np.arange(samples_per_chip), metrics, linewidth=1.2, color=line_color)
axes[5].set_xlabel("Code phase [samples]")
axes[5].set_ylabel("|Corr| over 1 symbol")
axes[5].set_title(f"Estimated code phase = {est_code_phase}  (true offset = {code_phase_offset})")
beautify(axes[5])

# Recovered message vs original (symbol-level)
n = np.arange(num_symbols)
axes[6].step(n, msg_symbols, where='mid', linewidth=3, color=line_color, label="Original")
axes[6].step(n, msg_hat, where='mid', linewidth=1.5, color='white', label="Recovered")
axes[6].set_xlabel("Symbol index")
axes[6].set_ylabel("Bit (+1/−1)")
axes[6].legend(loc="upper right")
axes[6].set_title(f"BER = {ber:.3f}")
axes[6].legend(fancybox=True, framealpha=1, facecolor=(0.6, 0.61, 0.8))

beautify(axes[6])

plt.show()
```

**Explanation:**
- Each **chip** of the PRN code lasts about $1\mu s$, so the full sequence of 10230 chips, transmitted at 1.023 Mcps, repeats every **10ms**.
- The **BPSK-modulated** signal looks like a sine wave whose phase (sign) changes following the code.  
- The **correlation peak** occurs when the locally generated code aligns perfectly with the received one — the delay of this peak tells the receiver *how long* the signal traveled.

> Even with a weak and noisy signal, the correlation process lets the receiver “recognize” the satellite’s unique code pattern.
