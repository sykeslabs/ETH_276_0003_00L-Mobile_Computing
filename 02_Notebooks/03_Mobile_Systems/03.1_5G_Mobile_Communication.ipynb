### 1.1 Introduction

5G, or fifth-generation mobile technology, became the global standard for telecommunications networks with commercial launches in 2019. Like its predecessors 3G, 4G, and 4G LTE, it operates within the same broad range of radio frequencies, but advances in speed, latency, and bandwidth allow for faster downloads and uploads, stronger connectivity, and more reliable service. These improvements make 5G the natural successor to 4G technology.

Not all of the ambitious goals set before 2017 have been achieved in practice. Targets such as one-millisecond latency, full-duplex transmission, and widespread use of millimeter wave have not been realized in everyday networks. Instead, real-world deployments have been shaped by trade-offs in spectrum availability, cost, and coverage. Research has already begun to look beyond 5G toward 6G, which is expected to revisit some of these unrealized technologies, including true full duplex, terahertz frequencies, reconfigurable intelligent surfaces, and AI-driven network optimization.

The broader history of mobile communications can be seen as a progression of technological generations, each defined by internationally agreed standards that set new performance benchmarks and enable new applications.

**First Generation (1G, 1980s).** The earliest mobile networks were based on analog technology. They enabled basic voice communication but lacked encryption, making calls vulnerable to interception. Devices were physically large and coverage was geographically limited.

**Second Generation (2G, 1990s).** Phones switched to digital signals, which made calls clearer and more secure. This was also when text messaging (SMS) became popular.

**Third Generation (3G, 2000s).** Mobile internet arrived. Phones could now send emails, browse the web, and even make video calls. This was the beginning of the smartphone era.

**Fourth Generation (4G, 2010s).** Internet on phones became fast and reliable, similar to home broadband. People could stream YouTube or Netflix, join video calls, and use all kinds of apps on the go.

**Fifth Generation (5G, 2020s).** Today’s generation is not just about faster speeds. 5G is built for things like smart devices in homes and cities, self-driving cars, and virtual reality, all requiring quick and reliable connections.


### 1.2 Use Cases and Original Technical Requirements

#### 1.2.1 Use Cases

A number of use cases were debated during the early development process of 5G. Overall, three categories were identified:

 * Enhanced Mobile Broadband (eMBB): Enables very high-speed video streaming and immersive experiences such as virtual and augmented reality.
 * Ultra-Reliable Low-Latency Communication (URLLC): Essential for self-driving cars, industrial automation, remote surgery, and other mission-critical services where even a small delay can be dangerous.
 * Massive Machine-Type Communication (mMTC): Connects millions of everyday devices, from smart appliances in homes to city-wide sensor networks that support smart infrastructure.

#### 1.2.2 Technical Requirements

There are ambitious design goals for 5G. 5G networks are designed to deliver a step change in performance compared to earlier generations. They provide significantly higher throughput, with gigabit-level download and upload speeds becoming practical in real deployments. Latency is greatly reduced, often measuring under 10 milliseconds in everyday use and approaching the one millisecond mark under optimized conditions. Reliability is also enhanced, ensuring stable connections even when networks are heavily loaded. At the same time, the system architecture increases overall capacity to the point of supporting up to one million devices per square kilometer, enabling dense deployments of sensors, smartphones, and other connected devices. The following indicators can be aligned with ITU IMT-2020 goals [[04](#ref04)].

* Latency: ~1 ms
* Peak download speeds: ~20 Gb/s
* Connection density: ~1 million devices/km²
* Spectral efficiency: 3-4 × higher than LTE
* Energy efficiency: lower per-bit transmission cost

#### 1.3.2 Slightly Modified Air Interface

The physical and medium access control layers of 5G mobile systems are designed to achieve very high spectral efficiency, flexible resource allocation, and low latency while remaining compatible with diverse deployment environments. On the physical layer, the air interface is based on orthogonal frequency division multiplexing, which had already been proven in LTE. In 5G New Radio, the design introduces scalable numerologies, meaning that subcarrier spacing can be varied in multiples of 15 kHz to accommodate different frequency ranges and latency requirements. This allows operation both in low bands with long symbol durations and in millimeter wave bands with shorter symbol durations. Downlink transmissions employ orthogonal frequency division multiple access, enabling multiple users to share subcarriers efficiently, while the uplink can use either single carrier frequency division multiple access or OFDMA, depending on device capability and scenario.

For digital modulation, 5G retains quadrature phase shift keying (QPSK) and 16-QAM as baseline schemes but extends the maximum modulation order to 256-QAM for both uplink and downlink in favorable channel conditions. This higher order modulation increases data rates at the cost of stricter signal-to-noise (SNR) requirements. Adaptive modulation and coding are central to the design, with the base station dynamically adjusting the modulation scheme and code rate according to instantaneous channel quality reports from the terminal. This ensures that spectral resources are used efficiently while maintaining reliable links.

At the MAC layer, 5G emphasizes flexible frame structures and dynamic scheduling. The time division duplex (TDD) mode dominates in mid-band deployments, enabling downlink and uplink resources to be reallocated as traffic demands change. The MAC scheduler at the base station assigns resource blocks in both time and frequency domains, with finer granularity than in LTE, and supports mini-slots that allow very short transmission bursts. These features are essential for ultra-reliable low latency communication, where packets may need to be transmitted and acknowledged within milliseconds. The MAC also incorporates mechanisms for massive machine-type communications by supporting grant-free uplink transmissions and contention-based access schemes, which reduce signaling overhead for devices that transmit small amounts of data sporadically.

Together, the modulation schemes and MAC protocols in 5G form an adaptable air interface that can operate efficiently across a wide variety of propagation environments and service requirements. They combine established multicarrier modulation principles with innovations in frame design, scheduling, and coding to deliver the performance envisioned in the IMT-2020 specifications.


#### 1.3.3 New Radio Spectrum for 5G

5G mobile communication systems rely on a much broader span of the radio spectrum than previous generations, reflecting the need for higher data rates, lower latency, and massive device connectivity. Regulators and standards bodies generally group the usable spectrum into two ranges. The first is the so-called FR1 or sub-6 GHz range, which extends roughly from 410 MHz up to 7.125 GHz. Within this range, operators commonly use bands around 600 MHz, 700 MHz, 2.3 GHz, 2.5 GHz, 3.5 GHz, and 4.8 GHz. These frequencies are well understood from the 4G era and provide wide area coverage and indoor penetration, but they have limited bandwidth, typically up to about 100 MHz per carrier, which constrains peak throughput.

The second major range is FR2, often referred to as millimeter wave, covering frequencies from 24.25 GHz to about 71 GHz. Bands that have seen early deployment include 26, 28, 39, and 47 GHz. These bands offer very large contiguous bandwidths, often 400 MHz or more, which enables multi-gigabit per second data rates. However, their propagation is limited, with signals strongly attenuated by walls, foliage, and even rain, which requires dense deployment of base stations or small cells in urban settings.

>The overall 5G spectrum strategy is therefore heterogeneous. Low-band frequencies below 1 GHz are used primarily for wide coverage and rural connectivity. Mid-band frequencies between 1 and 6 GHz provide a balance between capacity and coverage and are widely regarded as the workhorse of commercial 5G networks. High-band millimeter wave frequencies are deployed in hotspots, stadiums, and dense city centers where extreme capacity is required. Together, these bands allow 5G to address diverse service requirements, from enhanced mobile broadband to massive machine-type communications and ultra-reliable low latency applications.

International Telecommunication Union specifications under the IMT-2020 framework define these ranges, but national regulators allocate specific portions depending on local spectrum policy. As a result, the exact frequencies used for 5G vary across regions, though the overall division into low, mid, and high bands remains consistent worldwide.

#### 1.3.4 Unlicensed Spectrum for 5G

5G was designed primarily for operation in licensed spectrum allocated by national regulators, but the standard also includes mechanisms for unlicensed use. With Release 16, 3GPP introduced New Radio–Unlicensed (NR-U), which allows 5G to function entirely in unlicensed bands such as 5 GHz and 6 GHz, provided it complies with regional rules like listen-before-talk protocols, duty cycle limits, and power restrictions. A related approach, Licensed-Assisted Access (LAA), first developed for LTE, aggregates unlicensed spectrum with a licensed anchor channel to increase capacity, and this method is also available in 5G. In addition, some countries have adopted shared spectrum frameworks, such as the Citizens Broadband Radio Service at 3.5 GHz in the United States, where 5G can coexist under a tiered access model.
> A tiered access model is a way to share spectrum by giving different groups of users priority levels. In a specific band at 3.5 GHz, for example, in the U.S., military radar has top priority, licensed operators hold mid-tier priority through paid licenses, and general users access the lowest tier on an opportunistic basis. A central Spectrum Access System coordinates use so higher tiers are always protected, while unused spectrum can still be made available to others.


While these options broaden the usable spectrum and reduce dependency on licensed allocations, most commercial 5G deployments today continue to rely on licensed bands because they provide more predictable performance and tighter control over interference.


### 1.4 Critical Remarks

As summarized [[02](#ref02)] and [[03](#ref03)], 5G promised a lot of new developments.

5G is designed to deliver much higher data rates, far lower latency, and vastly increased capacity compared with LTE/4G. The targets are ambitious: latency below one millisecond (compared with around 70 ms for 4G) and peak download speeds of up to 20 Gb/s. Beyond faster mobile broadband, 5G has also been seen as an enabler for emerging technologies such as autonomous vehicles, virtual and augmented reality, and the Internet of Things.

> Rolling out 5G at scale, however, presents significant challenges. High-performance coverage requires a dense network of base stations, which is costly to build and maintain, while higher-frequency signals suffer from limited range and poor indoor penetration. Many devices are not yet 5G-capable, slowing user adoption. Networks must also remain interoperable with 4G and even 3G, since phones routinely switch between generations depending on coverage, adding technical and deployment complexity. Energy demand is another concern, with higher consumption raising both economic and environmental issues. At the same time, the vast number of connected devices expands the surface area for security and privacy risks.

It remains uncertain which combination of technologies will dominate in real deployments, as trade-offs in cost, complexity, and energy efficiency continue to shape the design of 5G networks. Coverage in rural and sparsely populated areas is especially difficult, given the need for more base stations and the limited propagation of high-frequency signals. Technical barriers such as interference management, advanced signal processing, and hardware challenges—particularly in realizing full duplex—also remain unresolved.
