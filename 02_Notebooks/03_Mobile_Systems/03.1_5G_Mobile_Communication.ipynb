### 1.1 Introduction

5G is the fifth generation of cellular mobile communication. It combines a radio interface called New Radio (NR) with network functions designed for broadband access, time-sensitive communication, and large numbers of connected devices. The capabilities available to an application depend on the spectrum, equipment, network configuration, and radio conditions.

The generations provide a useful historical overview. 1G introduced analog mobile voice. 2G brought digital voice and text messaging; 3G expanded mobile data access. 4G made mobile broadband central to the network design. 5G extends these capabilities and supports a wider range of service requirements.

This notebook introduces the main 5G use cases, explains the radio technologies behind them, and distinguishes design targets from application performance. Early papers such as *What Will 5G Be?* and *Five Disruptive Technology Directions for 5G* provide context for the original proposals [\[06\]](#ref06), [\[07\]](#ref07). The questions at the end connect spectrum, modulation, and network design to practical choices.


### 1.2 Use Cases and Original Technical Requirements

#### 1.2.1 Use Cases

The IMT-2020 framework groups 5G use cases into three broad categories [\[04\]](#ref04).

Enhanced Mobile Broadband (eMBB) emphasizes data rate and capacity for applications such as video streaming and large downloads. Ultra-Reliable Low-Latency Communication (URLLC) emphasizes timely, dependable delivery, for example in industrial control. Massive Machine-Type Communication (mMTC) emphasizes connecting many devices that often send small amounts of data, such as meters and environmental sensors.

These categories emphasize different priorities. A battery-powered sensor may need years of operation and reliable coverage, while a video application needs sustained throughput. The IoT background paper discusses this broader range of device and service requirements [\[08\]](#ref08).

#### 1.2.2 Technical Requirements

ITU's IMT-2020 evaluation requirements include the following values [\[10\]](#ref10). Each belongs to a defined evaluation scenario; they are not simultaneous guarantees for every connection.

| Metric | Requirement and context |
|---|---|
| Peak data rate | 20 Gbit/s downlink and 10 Gbit/s uplink under ideal conditions |
| User-experienced data rate | 100 Mbit/s downlink and 50 Mbit/s uplink in the specified dense urban eMBB evaluation |
| User-plane latency | 4 ms for eMBB and 1 ms for URLLC under the specified radio-interface conditions |
| Connection density | 1 million devices/km² for the specified mMTC traffic model |

User-plane latency concerns delivery of user data across the radio interface. An application's end-to-end delay also includes transport through the rest of the network and processing at the server. Similarly, peak radio capacity differs from the throughput of a download shared with other users.


### 1.3 5G Technology

#### 1.3.1 Core Innovations

Several approaches can increase capacity or improve coverage. Their benefits depend on where and how they are used. The early research papers discuss these design choices and their trade-offs [\[06\]](#ref06), [\[07\]](#ref07).

| Technology | Main idea | Practical limitation |
|---|---|---|
| Higher-frequency spectrum | Use wider available frequency bands to carry more data. | Blockage and propagation losses complicate coverage. |
| Small cells | Reuse spectrum over smaller areas and bring base stations closer to users. | More sites need installation, power, and network connections. |
| Massive MIMO | Use many antennas to separate and serve multiple data streams. MIMO means multiple-input multiple-output. | Gains depend on channel conditions and accurate channel estimates. |
| Beamforming | Adjust antenna signals to concentrate energy in useful directions. | Beams need to follow users and changing propagation paths. |
| Same-frequency full duplex | Transmit and receive simultaneously on the same frequency. | A transmitter's own signal can overwhelm its receiver unless sufficiently suppressed. |

5G also includes network-level changes. Network slicing creates logical networks on shared infrastructure for services with different requirements. The background paper on slicing explains how software-defined networking and virtualized network functions support this approach [\[09\]](#ref09).

#### 1.3.2 The 5G Radio Interface

NR uses orthogonal frequency division multiplexing (OFDM): data travels over many closely spaced subcarriers. Its numerology specifies their spacing and the associated symbol timing. Spacings such as 15, 30, and 60 kHz let the same radio design accommodate different operating conditions [\[11\]](#ref11).

Adaptive modulation and coding matches each transmission to channel quality. QPSK carries fewer bits per symbol but tolerates more noise; 16-QAM and 256-QAM carry more bits when reception is sufficiently reliable. Error-correcting codes add protection. Higher modulation orders therefore help only when the channel supports them.

The scheduler assigns time and frequency resources to devices. Mini-slots allow transmissions shorter than a full slot, reducing scheduling delay. Configured grants reserve uplink opportunities in advance, avoiding a separate scheduling request for each packet; this is often called grant-free access [\[11\]](#ref11).

Uplink means device to base station; downlink means the reverse. Time division duplex (TDD) separates these directions in time. Frequency division duplex (FDD) uses separate frequency bands. Both differ from simultaneous transmission and reception on the same frequency [\[12\]](#ref12).

#### 1.3.3 Radio Spectrum for 5G

3GPP defines the following NR frequency ranges in TS 38.104, Release 18 [\[12\]](#ref12). These ranges contain individual operating bands; a device or network supports only a selection of them.

| Range | Frequency limits |
|---|---|
| FR1 | 410 MHz to 7.125 GHz |
| FR2-1 | 24.25 to 52.6 GHz |
| FR2-2 | 52.6 to 71 GHz |

FR1 is sometimes called “sub-6 GHz,” although its upper limit exceeds 6 GHz. FR2 is commonly associated with millimeter-wave communication, although the strict millimeter-wave range starts at 30 GHz.

Lower frequencies generally suit broad coverage and indoor reach. Mid-band spectrum offers a compromise between coverage and bandwidth. Higher bands can provide large bandwidths over smaller areas, but obstacles and beam alignment become more significant. Antennas, transmit power, and the environment also affect the achievable range.

National regulators determine which bands may be used and under what conditions. The 3GPP ranges define equipment requirements; they do not themselves authorize spectrum use.

#### 1.3.4 Unlicensed and Shared Spectrum

NR-Unlicensed (NR-U), introduced in Release 16, supports operation in unlicensed spectrum, including deployments with a licensed anchor and standalone operation. Access remains subject to regional coexistence rules. Where listen-before-talk is required, a transmitter checks whether the channel is available before using it [\[13\]](#ref13).

Shared spectrum can also assign priorities. The United States' Citizens Broadband Radio Service (CBRS) has three tiers: incumbents receive the highest protection, Priority Access License holders have priority over general users, and General Authorized Access users must yield to both. A Spectrum Access System coordinates access [\[14\]](#ref14).

Licensed, unlicensed, and tiered shared access offer different levels of control over interference and access to capacity. Choosing between them depends on the service requirements and local spectrum rules.


### 1.4 Interpreting 5G Performance

A useful comparison identifies what was measured, under which conditions, and across which part of the system. The label “5G” alone does not specify a download speed, an application delay, or a reliability level.

#### 1.4.1 Reading Performance Metrics

| Metric | What a comparison needs to specify |
|---|---|
| Throughput | Available bandwidth, number of active users, signal quality, and whether the value is a peak or a sustained measurement |
| Latency | One-way delay or round-trip time, the endpoints, and the contribution of scheduling, retransmissions, and server processing |
| Reliability | Packet size, delivery deadline, loss criterion, and whether retries are included |
| Connection density | How often each device transmits and how much data it sends |
| Energy efficiency | Energy per delivered bit or total system energy, including the stated equipment and traffic load |

A short radio-interface delay can coexist with a much longer application response time. A high peak data rate can coexist with modest throughput during a busy period. For an example based on measured systems, the repository includes a comparison of commercial WLAN and 5G LAN equipment [\[15\]](#ref15). Its results apply to the tested setup.

#### 1.4.2 Relating Technology to Deployment

The technologies in section 1.3 solve different problems. Extra spectrum and multiple antenna streams increase potential capacity; small cells improve local reuse and coverage. Each also requires suitable equipment, planning, and supporting network connections. A feature's inclusion in a standard does not mean it is enabled in every deployment.

Same-frequency full duplex should be distinguished from FDD, where a device can transmit and receive on separate frequencies. Similarly, support for low-latency services should be distinguished from a guarantee for every application. These distinctions make comparisons between early 5G proposals, standards, and measurements more useful.

ITU's IMT-2030 framework describes objectives for the next generation, commonly called 6G [\[05\]](#ref05). Those objectives provide research direction; they should be assessed separately from demonstrated results.

#### 1.4.3 Looking Ahead: 6G and Artificial Intelligence

Artificial intelligence (AI) features prominently in 6G visions, but the ambition is not yet precise. The term “AI-native” suggests that learning-based functions should be built into the network from the outset. It leaves open which functions need AI, what should replace established methods, and how success should be measured. Samsung Research's 2025 white paper, *AI-Native & Sustainable Communication*, illustrates this direction and its promises of improved performance, automation, and energy efficiency [\[16\]](#ref16).

Some proposals are concrete enough to investigate. A learned encoder and decoder could reduce the amount of channel-state information sent from a device to a base station. Traffic prediction could help decide when lightly loaded cells can sleep. These examples define specific problems, but they do not by themselves establish the need for a new mobile generation or show that AI is the best solution [\[16\]](#ref16).

The broader direction remains uncertain: a collection of promising AI applications does not yet amount to a clearly justified network architecture. Claims of greater efficiency need comparisons with strong conventional methods, including the costs of data collection, training, inference, and model updates. Reliability under unfamiliar conditions and the consequences of incorrect predictions also need attention.

This uncertainty makes the topic worth investigating academically. Useful research turns broad promises into testable questions: under which conditions does a particular AI method help, how large is the benefit, and when does it fail? The value lies in establishing those limits as carefully as demonstrating improvements.

At ETH Zurich, several research activities address concrete parts of this agenda. Christoph Studer's Integrated Information Processing (IIP) group combines wireless communication, machine learning, and efficient hardware. Its experimental 5G platform supports work on ML-assisted signal detection and decoding, and positioning from radio signals. Such experiments help test whether proposed algorithms work beyond simulation [\[17\]](#ref17).

Hua Wang's Integrated Devices, Electronics, and Systems (IDEAS) group approaches the topic through radio hardware, including millimeter-wave and terahertz circuits and AI-assisted adaptation of radio systems [\[18\]](#ref18). The SwissChips 6G Communications programme covers a broader set of enabling technologies, including multi-antenna systems, energy-efficient circuits, sensing, and positioning [\[19\]](#ref19). These activities investigate specific components and trade-offs within the broader 6G vision.
