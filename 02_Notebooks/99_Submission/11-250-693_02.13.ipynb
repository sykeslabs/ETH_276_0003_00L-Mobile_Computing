### 13.1 From Decoding to a Live Link

A communication system must do more than recover data from an ideal signal. It must deliver a message through a physical environment, where distance, background noise, and other transmitters affect reception. An acoustic link makes these challenges directly observable: the transmission can be heard, its waveform can be inspected, and its success can be checked on a second device. The activity brings together modulation, noise, synchronization, and framing in a working system.

In [notebook 02.12](02.12_Acoustic_Synchronization_and_Framing.ipynb), a complete receiver recovered text from a recorded signal. This notebook takes the next step by exchanging messages live between two devices using frequency-shift keying (FSK):

$$\text{text} \rightarrow \text{FSK frame} \rightarrow \text{loudspeaker} \rightarrow \text{room} \rightarrow \text{microphone} \rightarrow \text{decoded text}.$$

The web application needs an internet connection to load. Once it is open, messages travel through the room as sound, using the devices' loudspeakers and microphones. Changing the settings and conditions makes it possible to investigate what allows a message to arrive reliably and what causes a transmission to fail.

The receiver decodes the tones according to the shared protocol. The messages are neither encrypted nor authenticated, so another nearby receiver with compatible settings may also recover them.


### 13.2 Open the Acoustic Link

Work in pairs and open **[audio-com.vercel.app](https://audio-com.vercel.app/)** on both devices. Begin with a short exchange under favourable conditions:

1. Select the same **Modulation** and **Speed** on both devices. Start with **8-FSK** and **Slow**.
2. Press **Listen** on both devices and allow microphone access.
3. Place the devices about half a metre apart and use a moderate playback volume.
4. Enter a name and a short message on one device, then press **Send**.
5. Check that the message appears on the receiving device, then send a reply. Both devices can remain in listening mode throughout the exchange.

The **Sent** status confirms only that the sender has finished playing the sound. Successful reception must be checked on the other device. A reply establishes that the link works in both directions.


### 13.3 Watch the Signal

The three plots present different views of the signal captured by the microphone. The **waveform** shows how its amplitude changes over a short time interval, making the start and end of a burst of sound visible. The **spectrum** shows the frequency content of the current analysis window. Its frequency markers indicate the candidate FSK tones for the selected modulation.

The **spectrogram** adds a time history, showing how the frequency content changes over the last few seconds. The data tones appear as short horizontal bands, while the opening and closing chirps sweep across frequencies. This view connects the live transmission to the frame structure examined in notebook 02.12.

Send a message and compare all three plots. They show the captured sound even when the receiver cannot decode it. Seeing a transmission in the plots and recovering its text are therefore separate observations.


### 13.4 Configure the Link

After establishing a working link, choose a modulation and speed for the pair and exchange two new messages. Agree on the settings before transmitting so that both receivers use the same tone set and symbol duration. The name entered in the application becomes the first part of the transmitted text, for example `Ada: Hello!`.

To compare settings, stop listening on both devices, change **one setting on both devices**, and press **Listen** again. Repeat the same message, keeping the name, distance, and volume unchanged. Compare the displayed bit rate, the duration of the transmission in the spectrogram, and whether the receiving device recovers the message.

Matching settings are essential because they define how the receiver interprets the signal. A receiver using the wrong tone set or symbol duration will usually fail to recover the text, even when the sound is clearly audible.


### 13.5 Challenge the Link

Choose **one** condition to investigate while keeping the others fixed. To explore the effect of **distance**, gradually move the devices farther apart until a message no longer arrives. To explore **interference**, introduce background sound near the receiver and repeat the transmission.

Alternatively, compare the same message with **Slow** and **Fast** settings, or compare **BFSK** with **8-FSK** at the same speed setting. Apply each setting to both devices and record how it affects the transmission duration and successful reception.

After a failure, change one condition and try again. For example, reduce the distance or remove the added background sound. Record both the failed attempt and the adjustment so that the result shows which change helped the link recover. Repeating a transmission under the same conditions can also reveal whether reception is consistently reliable or varies between attempts.


### 13.6 A Shared Medium

Sound spreads through the room, so a transmission can reach several microphones. If two devices transmit at the same time, their signals overlap at the receiver. The result may be a failed decoding attempt, or the receiver may recover one message, for example if one signal is much stronger. This makes coordination part of the communication problem: nearby pairs need a way to decide when to transmit.

If the class setup permits, arrange for two senders to transmit at nearly the same time while a third device listens. Observe the plots and record what the receiver reports. Compare this with transmissions sent separately; overlapping signals do not necessarily produce the same outcome on every attempt. This experiment introduces the shared-medium problem examined later in the [Wi-Fi contention protocol](../03_Mobile_Systems/03.5_Wi-Fi_Contention_Protocol.ipynb) notebook.


### 13.7 If a Message Does Not Arrive

Begin by checking that both devices are in listening mode, have microphone permission, and use matching modulation and speed settings. Keep the devices close together, use an audible playback volume, and allow only one sender to transmit at a time. These conditions provide a useful starting point for restoring a working link.

Use the live plots to distinguish between possible causes. If a transmission produces no visible change, check microphone capture, playback volume, and distance. If the signal is visible but no text appears, check the settings and look for noise or overlapping transmissions. A visible signal alone does not guarantee that its quality or timing is sufficient for decoding. Change one condition at a time and repeat the same message to assess its effect.

**Why should nearby pairs take turns transmitting? If overlapping transmissions were tested, describe what the receiver reported and compare the result with separate transmissions.**

Sound spreads through the whole room, so every microphone nearby hears every transmission. If two devices send at the same time, their tones mix at the receiver. The receiver can't tell which tone belongs to which message, so usually neither message is decoded correctly. Sometimes the stronger message still gets through. Taking turns avoids this, just as Wi-Fi devices have to share the same radio channel.
