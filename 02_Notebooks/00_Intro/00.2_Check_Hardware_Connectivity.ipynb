### 2.2 Raspberry Pi Setup

Use this section to check whether your machine can communicate with a Raspberry Pi over the network.

Check whether the Raspberry Pi is reachable. The next cell sends one ping request to the configured hostname and reports whether it receives an answer.

```python
raspberry_pi_conf = {
    "hostname": "raspberrypi.local",
    "username": "cas",
    "password": "password"
}
```

```python
import subprocess
import platform

param = "-n" if platform.system().lower() == "windows" else "-c"
command = ["ping", param, "1", raspberry_pi_conf["hostname"]]

try:
    output = subprocess.check_output(command, stderr=subprocess.STDOUT, universal_newlines=True)
    print(f"✅ {raspberry_pi_conf['hostname']} is reachable")
except subprocess.CalledProcessError:
    print(f"❌ {raspberry_pi_conf['hostname']} is NOT reachable")
```

Now check whether the Raspberry Pi is set up correctly by running a simple script on the device over SSH.

```python
import paramiko

command = "~/gettingStarted/helloPi.sh"

client = paramiko.SSHClient()
client.set_missing_host_key_policy(paramiko.AutoAddPolicy())
try:
    client.connect(**raspberry_pi_conf)
    stdin, stdout, stderr = client.exec_command(command)
    output = stdout.read().decode().strip()
    error = stderr.read().decode().strip()
    if output == "Hello World":
        print(f"✅ Script ran and output matches: {output}")
    else:
        print(f"❌ Unexpected output:\n{output}\nError:\n{error}")
finally:
    client.close()
```

### 2.3 Arduino

With the Arduino, we cannot simply ping the device. Instead, we use a simple ping-pong response over the serial connection. First, find the correct serial port:

- Windows
  1. Open the device manager.
  2. Expand the ports section.
  3. Look for a device named `Arduino`.
  4. The `COM` number, for example `COM3`, is your serial port string.

- macOS
  1. Open a terminal.
  2. Run `ls /dev/cu.usbmodem* /dev/cu.usbserial*`.
  3. Look for an entry similar to `/dev/cu.usbmodem14101`; this is your serial port string.

If you are unsure at any step, disconnect the Arduino and compare which serial device disappears.

```python
arduino_conf = {
    "port": "<insert serial port string>",
    "baudrate": 9600,
}
```

```python
import serial
import time

serial = serial.Serial(**arduino_conf, timeout=1)
time.sleep(2)

serial.write(b'ping\n')
print(f"Sending 'ping' to arduino on port {arduino_conf['port']}")
response = serial.readline().decode('utf-8').strip()

if response == "pong":
    print(f"✅ Arduino is reachable and the response matches: {response}")
else:
    print(f"❌ Unexpected output:\n{response}")

serial.close()
```
