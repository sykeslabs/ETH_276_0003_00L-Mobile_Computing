### 2.2 Raspberry Pi Setup

Use this section to check whether your machine can communicate with the course Raspberry Pi over the network.

Plug the Pi into power and connect it to your laptop with the Ethernet cable. Wait about a minute until the green light stops flickering: the Pi needs that long to boot. The Pi always has the same address, `172.31.250.1`. How this works is explained in [`05.04_GettingStarted_with_Raspberry_Pi.ipynb`](../05_Mobile_AI/05.04_GettingStarted_with_Raspberry_Pi.ipynb).

Check whether the Raspberry Pi is reachable and set up correctly. The next cell asks the course server on the Pi for its status report over HTTP and prints who answered.

```python
import requests

url = "http://172.31.250.1:8000/health"
try:
    health = requests.get(url, timeout=10).json()
    print(f"✅ {health['hostname']} answers on {url}")
    print(f"   Ollama running: {health['ollama']['reachable']}")
except requests.RequestException as error:
    print(f"❌ No answer from {url} ({type(error).__name__})")
```

### 2.3 Arduino

Unlike the Raspberry Pi, the Arduino has no network connection. Instead, we use a simple ping-pong response over the serial connection. First, find the correct serial port:

- Windows
  1. Open the device manager.
  2. Expand the ports section.
  3. Look for a device named `Arduino`.
  4. The `COM` number, for example `COM3`, is your serial port string.

- macOS
  1. Open a terminal.
  2. Run `ls /dev/cu.usbmodem* /dev/cu.usbserial*`.
  3. Look for an entry similar to `/dev/cu.usbmodem14101`; this is your serial port string.

If you are unsure at any step, disconnect the Arduino and compare which serial device disappears.

```python
arduino_conf = {
    "port": "<insert serial port string>",
    "baudrate": 9600,
}
```

```python
import serial
import time

serial = serial.Serial(**arduino_conf, timeout=1)
time.sleep(2)

serial.write(b'ping\n')
print(f"Sending 'ping' to arduino on port {arduino_conf['port']}")
response = serial.readline().decode('utf-8').strip()

if response == "pong":
    print(f"✅ Arduino is reachable and the response matches: {response}")
else:
    print(f"❌ Unexpected output:\n{response}")

serial.close()
```
