### 4.1 Introduction

This notebook introduces a practical workflow for using A.I. coding assistants in Visual Studio Code (VS Code). The goal is to use an assistant as a fast collaborator while you remain responsible for the requirements, the design choices, the code, and the verification.

Vibe coding means using an A.I. assistant to move quickly from intent to working code. Done well, this can speed up exploration, debugging, and implementation. Done carelessly, it can produce code that looks plausible but is untested, over-engineered, insecure, or unrelated to the original task.

### 4.2 Four Clauses for A.I.-Assisted Coding

The following four clauses are attributed in the linked Reddit discussion to Andrej Karpathy's [`CLAUDE.md`](http://claude.md/) prompt pattern: [Reddit discussion](https://www.reddit.com/r/ClaudeAI/comments/1uc7izy/i_added_a_clause_to_andrej_karpathys_4_claudemd/?share_id=xlf8YTyHSAM3Ss8UNZMuX&utm_content=2&utm_medium=android_app&utm_name=androidcss&utm_source=share&utm_term=2). They are useful as a compact operating contract for an A.I. coding assistant.

<pre><code>Ask, don't assume. If something is unclear, ask before writing a single line. Never make silent assumptions about intent, architecture, or requirements.
</code>
<code>Simplest solution first. Always implement the simplest thing that could work. Do not add abstractions or flexibility that weren't explicitly requested.
</code>
<code>Don't touch unrelated code. If a file or function is not directly part of the current task, do not modify it, even if you think it could be improved.
</code>
<code>Flag uncertainty explicitly. If you are not confident about an approach or technical detail, say so before proceeding. Confidence without certainty causes more damage than admitting a gap.</code></pre>

### 4.3 What to Check Before Accepting A.I.-Generated Code

Before accepting generated changes, inspect them in VS Code and ask yourself:

* Did the assistant change only the files needed for the task?
* Can I explain every changed line at a high level?
* Did it introduce unnecessary abstractions, dependencies, or configuration?
* Does the code handle the expected inputs and obvious edge cases?
* Did I run the relevant notebook cells, tests, or commands?
* Did the assistant state any assumptions or uncertainty?
* Is there any private information, secret, token, or personal data in the prompt or generated code?

A useful rule: if you cannot explain the generated code, you should not submit it as your own work.

### 4.4 Common Failure Modes

A.I. coding assistants often fail in predictable ways:

* **Over-building:** adding classes, configuration, or frameworks for a small task.
* **Silent assumptions:** choosing an architecture or interpretation without asking.
* **Scope drift:** editing nearby code that was not part of the task.
* **Plausible but wrong APIs:** inventing function names, package behavior, or command flags.
* **Weak verification:** claiming success without running the relevant checks.
* **Style mismatch:** using a different coding style than the surrounding project.
* **Security mistakes:** logging secrets, weakening validation, or mishandling private data.

The four clauses do not eliminate these failures, but they make them easier to detect early.
