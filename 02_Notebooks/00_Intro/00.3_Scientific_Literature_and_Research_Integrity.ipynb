### 3.2 Use of AI Tools

AI-based literature tools can help you explore a research area, discover related papers, and get an overview of a topic more quickly. They are useful for identifying potentially relevant sources, but they do not replace reading the original papers or checking whether a source is peer-reviewed, relevant, and correctly cited. When using such tools, follow ETH Zurich's guidance on the responsible use of generative AI in education: [Use of Generative AI, ETH Zurich](https://ethz.ch/en/the-eth-zurich/education/ai-in-education.html).

Useful tools for literature exploration include:

* [Connected Papers](https://library.ethz.ch/en/locations-and-media/media-types/databases-standards-patents/connected-papers.html): creates a graph of related publications starting from one known paper. This is useful when you have already found one relevant article and want to find related research.
* [Elicit](https://elicit.com/): lets you search for literature using research questions and can summarize the papers it finds. This is useful for quickly screening many papers, but you still need to check the summaries against the original papers.
* [ResearchRabbit](https://www.researchrabbit.ai/): helps organize paper collections and find related authors, papers, and research networks. This is useful for following a topic over time.
* [Scite](https://scite.ai/): shows how publications are cited by later work. The citation context can help you assess how later studies discuss or evaluate a paper’s findings.



Start with a manual search through the ETH Library, IEEE Xplore, or the ACM Digital Library. Then use an AI tool to find additional papers and compare the results. Cite the original publications using a consistent citation style, and check each reference for accuracy. Do not cite an AI-generated summary as if it were a scientific publication.


### 3.3 Primary, Secondary, and Tertiary Sources

Scientific writing distinguishes between primary, secondary, and tertiary sources.

* **Primary sources** present original research or original data. In this course, examples include peer-reviewed conference papers, journal articles, technical standards, datasets, and official technical reports.
* **Secondary sources** interpret, compare, or summarize primary sources. Examples include survey papers, review articles, textbooks, and technical overview articles.
* **Tertiary sources** collect and summarize established knowledge from many other sources. Examples include encyclopedias, dictionaries, handbooks, and Wikipedia pages.

Wikipedia can be useful at the beginning of a literature search because it often gives a readable overview, introduces key terms, and points to references that may be worth checking. For scientific claims in academic writing, you should normally cite the underlying research rather than a Wikipedia page. Use Wikipedia to identify keywords, standards, researchers, organizations, and references, then trace those references back to primary or secondary sources that you can cite directly.

### 3.4 Scientific Integrity and Good Scientific Practice

Scientific integrity is essential for producing reliable and trustworthy knowledge. ETH Zurich bases good scientific practice on four principles: reliability, honesty, respect, and responsibility. These principles apply to everyone involved in research and teaching at ETH Zurich.

In practice, scientific integrity means reporting methods and results honestly, documenting research so that it can be understood and verified, handling data carefully, acknowledging all sources, assigning authorship fairly, disclosing conflicts of interest, and treating collaborators and research participants responsibly. Making up data, altering data to misrepresent results, concealing relevant information, plagiarizing, and misrepresenting the contributions of others are forms of scientific misconduct.

These responsibilities also apply when using AI tools. AI-generated claims, summaries, and references must be checked against reliable original sources. The use of AI must be acknowledged when required by the course, supervisor, publisher, or applicable ETH guidance. Before including a reference suggested by an AI tool, verify that the publication exists and supports the claim you are making.



For further information, consult ETH Zurich’s [Scientific integrity](https://ethz.ch/staffnet/en/finance-and-controlling/compliance/wissenschaftliche-integritaet.html) overview and the [ETH Zurich Guidelines on Scientific Integrity (RSETHZ 414)](https://ethz.ch/content/dam/ethz/associates/ethlibrary-dam/documents/Aktuell/News/2022/ETH%20Zurich_Guidelines%20_on%20_Scientific%20_Integrity.pdf).

### 3.5 Citation Styles

A citation style defines how sources are presented both in the text and in the bibliography. Common styles include APA, Chicago, Harvard, IEEE, and MLA. Different academic fields use different styles, so always follow the style required by your course, department, supervisor, publisher, or journal. If no style is specified for an assignment, ask your instructor which one to use.

The ETH Library distinguishes between author–date styles, numerical styles, and footnote styles. In computer science and engineering, the numerical IEEE style is commonly used: citations appear as numbers in square brackets in the text, and the full references are listed in the order in which they are first cited. Whichever style you choose, apply it consistently and verify automatically generated references against the original publication.

For explanations and examples, consult the section “What are citation styles?” on the ETH Library page [Prevention of plagiarism](https://library.ethz.ch/en/scientific-writing/prevention-of-plagiarism.html). The page also links to the ETH-licensed _Cite Them Right_ resource for detailed guidance on formatting different types of sources.
