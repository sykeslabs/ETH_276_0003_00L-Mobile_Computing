### 5.1 What Git Does

Git is a version-control system. It records changes to files, lets you compare versions, and helps several people work on the same project. A project tracked by Git is called a **repository**.

It is useful to distinguish three places:

1. The **working tree** contains the files currently on your computer.
2. The **staging area** contains changes selected for the next commit.
3. The **repository history** contains saved snapshots called commits.

A **remote** is another copy of the repository, usually hosted on a server such as ETH GitLab. In this course, the remote repository distributes notebooks, assignments, and updates.

### 5.2 Check Git on Your Computer

Open a terminal and run:

```bash
git --version
```

A version number confirms that Git is installed. If the command is not found, return to the Git installation instructions in [`README.md`](../../README.md).

Git commands normally affect the repository containing your terminal's current directory. Before running a command, check where you are:

```bash
pwd                 # macOS and Linux
Get-Location        # Windows PowerShell
```

### 5.3 Access the Course Repository

ETH hosts the course repository on GitLab. No account is required for read-only access. To download it for the first time, move to the directory in which you want to keep the course material and run:

```bash
git clone --depth 1 https://gitlab.ethz.ch/wireless/MobileComputing.git
cd MobileComputing
```

`git clone` creates a new local repository. The `--depth 1` option downloads only the newest version of the history, which is sufficient for accessing the course material. Do not clone the repository again if the `MobileComputing` directory already exists.

To check the address of the remote repository from inside an existing clone, run:

```bash
git remote -v
```

### 5.4 Inspect and Update Your Copy

Before updating, inspect your local repository:

```bash
git status
```

`git status` shows the current branch and any changed, staged, or untracked files. It is safe to run because it does not modify anything.

If the working tree is clean, download and integrate the newest course version with:

```bash
git pull
```

A successful update may end with `Fast-forward`. Warnings about notebook metadata do not necessarily mean that the pull failed; read the final lines and run `git status` again.

If `git status` lists your own changes, save a separate backup and inspect them with `git diff` before pulling. Do not discard files or use destructive recovery commands unless you understand what they will remove. Ask the teaching team if Git reports a merge conflict.

### 5.5 Record Changes in a Repository

The usual local workflow is:

```bash
git status
git diff
git add path/to/file
git commit -m "Describe the change"
```

`git add` selects a change for the next commit. `git commit` records the selected changes in your local repository. A clear commit message briefly explains what changed.

The command `git push` sends local commits to a remote repository, but it requires write permission and authentication. Anonymous access to the course repository allows you to clone and pull; it does not grant permission to push. Follow the course submission guidelines instead of trying to push your completed notebooks to the course repository.

### 5.6 Branches and Interactive Practice

A **branch** is a movable name for a line of development. Branches let you try changes without immediately adding them to the main line. Useful commands include:

```bash
git branch                 # list local branches
git switch -c my-practice # create and switch to a new branch
git switch main           # switch to an existing branch named main
```

Branch names differ between repositories, so check `git status` or `git branch` before switching.

The free [Learn Git Branching](https://learngitbranching.js.org/) tutorial provides a visual practice repository in your browser. Start with the introductory levels on commits and branches. The tutorial is separate from the course repository, so experimenting there cannot damage your course files.

### 5.7 A Small Daily Checklist

When working with the course repository:

1. Open a terminal in the `MobileComputing` directory.
2. Run `git status`.
3. If the working tree is clean, run `git pull`.
4. Open the current notebook and complete your work.
5. Save your notebook and submit it according to the course submission guidelines.

For more detail, consult the official [Pro Git book](https://git-scm.com/book/en/v2), especially the chapters *Getting Started* and *Git Basics*.
