### 1.2 Introduction: From a Local Machine to the Cloud

The *cloud* appears in many different contexts. It may refer to online storage, video streaming and web applications. Instead on running on the user's device these services rely entirely on computing resources operated elsewhere, accessible through the network. 

However, not every remote computer is a cloud. This section aims to gradually go from a simple system - a Linux machine at home - to a cloud-based service. Each step comes with its own challenges and responsibilities. [1, 6]

#### 1.2.1 Step 1 - Linux Machine at Home

The Linux machine builds the starting point of a service. Assume you would like to offer a service that modifies images. the current machine provides the whole compute, storage and networking capabilities. Which bears the disadvantage that the users are fundamentally limited by the purchased hardware. If you now suddenly realise that the graphical processing unit is not powerful enough to get the desired result you would have to upgrade the hardware. At the same time, you, as the owner, gets the absolute control, as well as the responsibility over the setup. This includes being able to determine what happens with the data with a high certainty. At the same time you have to bear the burden of taking care of the security, backups, and software stacks. Upgrades, patches and changes in the software require knowledge and skill and any disruption, be it in software, power, or network failure, can stop the service. This setup has another significant disadvantage: at the moment, users of the image processing service would have to get to the Linux machine. While this makes the setup relatively secure, it also means that the potential users of the service is limited to the proximity of the machine.  [1, 16]

#### 1.2.2 Step 2 - Making the Machine Public

Now, you notice that the proximity is a severe bottleneck and you decide to make the machine publicly accessible. For this you ensure that your server has an static IP address allocated and you buy a domain that directly points to this address. Now, whenever anyone visits your domain it will fetch the static IP from the Domain Name Server (DNS). [8] 

While it allows anyone to access your service, this significantly enlarges the attack surface. Before, monitoring, patching and access control were somewhat optional, they become now mandatory. [16] Users can access your IP address, but, they don't know *how* to communicate with your service. Thus, you need to define endpoints with pre-determined request and response patterns, so that, interaction becomes possible. [9] Additionally, to protect the data which is sent you need to encrypt the traffic through TLS and authenticate the server. Otherwise, eavesdropping is possible or malicious actors may impersonate your server. [11] Both measures increases the complexity of your service and need to maintained to ensure that authentication, privacy and security is guaranteed. 



#### 1.2.3 Step 3 - Renting a Server

In both previous versions your service is still limited by the hardware you initially bought. So, you decide to rent a Virtual Private Server (VPS). Now, a provider takes care of the physical server, the power supply, cooling and the connection to the network. You don't have to deal with the maintenance of the hardware and whenever your VPS cannot keep up with the computational expectations you can switch to a more powerful one - all without having to invest money in new hardware. [6]

However, at any given moment you still rent one server with a fixed capacity. Switching to a more powerful VPS usually remains a manual decision and may require you to stop or restart the image processing service. You also still have to install and patch the operating system, configure the public endpoint, protect the server and maintain the application and its data. Thus, the location and ownership of the physical hardware changed, but many of your previous responsibilities and limitations remain. A rented server is therefore not automatically a cloud service. [1, 6] Additionally, you also lost some control. You may have the option to choose location in an abstracted way (such as specific data center or country) but you do not know the precise location of your data. 



#### 1.2.4 Step 4 - Moving to the Cloud

After operating the image processing service for some time, you notice that the number of requests is not constant. During some periods the server is mostly idle, while during others more images arrive than it can process. Renting an even larger VPS could handle the busy periods, but you would also pay for this additional capacity whenever it is not needed. Instead, you decide to move the service to the cloud.

In the cloud, computing resources can be provisioned and released on demand. Instead of permanently relying on one server, you can add more processing capacity when many images arrive and remove it again when demand decreases. The resources come from a shared pool and their usage can be measured, allowing you to pay according to the resources you consume. To you, the available resources may appear almost unlimited because you no longer have to purchase and install every additional machine yourself. [1, 3]

At the same time, you lose some of the control you had over the data. When the images were stored on the Linux machine at home, you knew on which physical device they were located. In the cloud, the resources are assigned dynamically and the data may be replicated across several systems. You may be able to select a country or region, but you generally do not know which exact physical server or storage device contains an image. [1, 16]

Consequently, you need to trust that the provider handles the images in the way you envision it. This includes who is allowed to access the data, whether it is encrypted, where backups are stored, how long copies are retained and whether the data is actually deleted when you request it. This trust should not only be based on the provider's promises. Contracts, documented responsibilities, security controls and independent assessments can help you determine whether the service fulfils your requirements. Nevertheless, you give up some direct control in exchange for the flexibility offered by the cloud. [6, 16]

This flexibility requires the provisioning and operation of resources to become automated. At the same time, moving to the cloud does not mean that all responsibilities disappear. Which parts are maintained by the provider and which parts remain with you depends on the chosen cloud service. The cloud therefore changes how the image processing service can be operated, scaled and paid for, but it also introduces new dependencies and decisions. [6]

**Question:** What distinguishes cloud computing from conventional remote hosting? [1, 6]



### 1.3 Cloud Computing Models

The previous section gradually moved the image processing service away from a local machine. However, the final step was simply described as moving to *the cloud*. Before this difference can be discussed, it is necessary to determine what the term cloud computing actually describes.

#### 1.3.1 Definitions

Cloud computing is not defined in exactly the same way everywhere. The NIST definition describes it as convenient, on-demand network access to a shared pool of configurable resources. It provides a particularly structured view by separating five essential characteristics, three service models and four deployment models. [1] ISO/IEC 22123-1 approaches the topic through a standardized vocabulary, with the goal of using cloud-related terms consistently across systems and standards. [2]

Other definitions place a stronger focus on the motivation behind cloud computing. The Berkeley View describes applications delivered as services together with the large datacenters that provide them. Its focus lies on elasticity, the ability to acquire and release resources quickly, and the economic advantage of paying for resources as they are used. [3] Vaquero et al. compare more than twenty existing definitions and identify virtualized, scalable and dynamically provisioned resources as recurring elements. [4] Buyya et al. describe cloud computing as a market-oriented utility in which resources are delivered according to service-level agreements and an expected quality of service. [5]

| Source | Main emphasis | Recurring elements |
|---|---|---|
| NIST SP 800-145 [1] | A structured cloud model | Self-service, network access, pooling, elasticity and measurement |  
| ISO/IEC 22123-1:2023 [2] | A common vocabulary | Consistent terminology for cloud concepts |  
| Berkeley View [3] | Technical and economic motivation | Elasticity, service delivery and pay-per-use |  
| Vaquero et al. [4] | A consensus across existing definitions | Virtualized, scalable and dynamically provisioned resources |  
| Buyya et al. [5] | Computing as a utility | Service-level agreements, quality of service and market-oriented allocation |  

Consequently, there is no single definition that is used in exactly the same way everywhere. Some sources describe the technical properties, others emphasize a common vocabulary, and others focus on the economic or service-oriented perspective. Nevertheless, several ideas are shared: resources are accessed through a network, provided as a service, pooled or shared, provisioned according to demand and scaled without purchasing new hardware for every change. Measurement, automation and a separation between the consumer and the provider also appear repeatedly. For the remainder of this notebook, NIST SP 800-145 is used as the working definition because it provides a clear structure for comparing the systems introduced in Section 1.2. [1, 2, 3, 4, 5, 6]

#### 1.3.2 Characteristics

The NIST definition describes five characteristics that need to be considered together. **On-demand self-service** means that computing resources can be provisioned when they are required without first interacting with an employee of the provider. **Broad network access** means that the service is available through standard network mechanisms and can be accessed by different types of clients. Making the Linux machine publicly accessible fulfilled parts of this characteristic, but did not yet make the machine a cloud. [1]

**Resource pooling** means that the provider's physical and virtual resources are shared between several consumers and assigned according to demand. The exact physical location is therefore usually abstracted. A region or country may be selected, but the consumer generally does not select the exact server. **Rapid elasticity** allows resources to be added and removed quickly. Finally, **measured service** means that the usage of resources, such as processing, storage or bandwidth, is monitored and reported. This can also form the basis for usage-dependent costs. [1]

These characteristics describe how a cloud behaves. They do not determine which parts of the image processing service are operated by the provider and which remain under the control of the consumer. This boundary is described through the service models.

#### 1.3.3 Service Models

The service models can be understood as different divisions of control and responsibility. From Software as a Service to Platform as a Service and finally Infrastructure as a Service, the consumer receives more control over the system. At the same time, the knowledge, skills and number of engineers required to operate it generally increase. The responsibility for security, maintenance and failure handling also moves increasingly towards the consumer. None of the models removes the need to trust the provider, as the physical infrastructure remains outside the consumer's direct control. [1, 6]

<p align="center"><img src="../../.admin/img/03_service_model_responsibilities.svg" alt="Responsibility stack comparing SaaS, PaaS and IaaS." width="78%"></p>

##### Software as a Service (SaaS)

With **Software as a Service**, a finished application is provided. In the image processing example, a consumer could open a website, upload an image, select one of the available modifications and download the result. The consumer can control the settings exposed by the application, but cannot change the image processing algorithm, install another library or decide how the underlying servers and storage are configured. [1]

The provider maintains the application, runtime, operating system and physical hardware. This removes the hardware investment and reduces the need for application developers, system administrators and infrastructure engineers on the consumer side. Instead, the service is paid for through a subscription or according to its usage. The consumer, however, has to trust that the provider processes, stores, backs up and deletes the images as expected. Responsibilities such as selecting appropriate settings, managing user accounts and deciding which data may be uploaded still remain with the consumer. [6, 16]

<details>
<summary><strong>Popular example: Google Workspace (SaaS)</strong></summary>

Google Workspace provides finished applications such as Gmail, Google Drive and Google Docs. Users can create content, share files and change the settings offered by the applications. They cannot select the operating system, replace the storage implementation or control the physical servers on which the applications run.

Further information: [Official Google Workspace website](https://workspace.google.com/)

</details>

##### Platform as a Service (PaaS)

With **Platform as a Service**, the application is developed by the consumer and deployed to a runtime provided by the cloud provider. For the image processing service, the processing algorithm, API, user interface and application-specific data handling can be implemented and changed. However, the application has to use the languages, libraries, interfaces and runtime versions supported by the platform. The operating system, physical servers and much of the networking remain outside the consumer's control. [1, 6]

PaaS therefore requires software engineers who can develop, secure and maintain the application, but it reduces the need for engineers who maintain operating systems and physical hardware. There are no direct hardware purchase costs, but the platform, storage and consumed resources still have to be paid for. The consumer is responsible for errors in the application, the authentication of its users and the way data is handled inside the application. At the same time, the provider has to be trusted to patch and isolate the runtime, keep the platform available and handle the underlying data according to the agreement. Provider-specific interfaces may also make it difficult to move the application to another platform. [6, 16]

<details>
<summary><strong>Popular example: Google App Engine (PaaS)</strong></summary>

Google App Engine allows developers to deploy an application to a managed environment. The application code and its behaviour remain under the developer's control, while the operating system, runtime infrastructure and physical servers are managed by the provider. The application must remain within the interfaces and runtime options supported by the platform.

Further information: [Official Google App Engine website](https://cloud.google.com/appengine)

</details>

##### Infrastructure as a Service (IaaS)

With **Infrastructure as a Service**, fundamental resources such as virtual machines, storage and networking are provisioned. The operator of the image processing service can select an operating system, install libraries, configure network rules and decide how the API, workers, storage and backups are structured. This provides substantially more freedom than PaaS or SaaS. However, the provider still controls the physical servers, storage devices, datacenter and the cloud infrastructure used to provide the virtual resources. [1]

This additional control also creates additional costs and responsibilities. Physical hardware does not have to be purchased, but virtual machines, storage, network traffic and supporting services have to be paid for. Engineers are required to configure and patch the operating systems, monitor the service, manage access, create backups and react to failures. More knowledge is therefore required to operate IaaS securely and reliably. Although the consumer can decide how many security measures are implemented inside the virtual infrastructure, the provider still has to be trusted to isolate consumers, protect the physical hardware and make the infrastructure available. [6, 16]

<details>
<summary><strong>Popular example: Amazon EC2 (IaaS)</strong></summary>

Amazon EC2 provides virtual machines with selectable processing capacity, memory, storage and networking. The operator can choose and configure the operating system and install the image processing software. The physical server, virtualization layer and datacenter remain under the control of the provider.

Further information: [Official Amazon EC2 website](https://aws.amazon.com/ec2/)

</details>

The progression from SaaS to PaaS to IaaS is therefore a trade-off. SaaS offers the least control but transfers most operational work to the provider. PaaS provides control over the application while the platform remains managed. IaaS provides control over most of the software stack, but also requires the most operational knowledge and responsibility. Greater control can reduce the amount of trust placed in the provider for some decisions, but it does not remove the provider from the system. [1, 6]

##### 1.3.1.3 Deployment Models

The service models describe which part of the system is provided. The deployment models instead describe who may use the cloud infrastructure and how it is organized. These two classifications are independent. For example, an application can be offered as SaaS from either a public or a private cloud. [1]

A **public cloud** is offered for open use and is operated by a cloud provider. Several unrelated consumers can use the provider's pooled resources. This can provide large economies of scale, but consumers have less direct control over the underlying infrastructure. A commercial image processing service running on the infrastructure of a public provider is one example. [1, 6]

<details>
<summary><strong>Popular example: Amazon Web Services (public cloud)</strong></summary>

Amazon Web Services offers computing, storage and application services to unrelated organizations and individuals. The resources are operated by the provider and made available to consumers through public cloud interfaces.

Further information: [Official Amazon Web Services website](https://aws.amazon.com/)

</details>

A **private cloud** is used exclusively by one organization. It may be located on the organization's premises or operated elsewhere by a third party. A university could, for example, operate a private cloud for its departments. This can provide more control over policies and data location, but the organization or its contractor also has to finance and maintain the required infrastructure and expertise. [1, 6]

<details>
<summary><strong>Popular example: An OpenStack private cloud</strong></summary>

An organization can use OpenStack to provide cloud resources exclusively to its own departments. The organization decides where the infrastructure is located and which internal users receive access, but it also has to operate or contract the underlying cloud environment.

Further information: [Official OpenStack website](https://www.openstack.org/software/)

</details>

A **community cloud** is used by several organizations with shared requirements, such as a common mission, security policy or legal obligation. Several universities could share such an infrastructure for research data. Costs and responsibilities can be shared, but common rules and a clear governance model are required. [1]

<details>
<summary><strong>Popular example: A government community cloud</strong></summary>

Government organizations can share a cloud environment that is restricted to institutions with common legal, security and compliance requirements. The infrastructure may still be operated by an external provider, but it is not offered for unrestricted public use.

A service such as [AWS GovCloud (US)](https://aws.amazon.com/govcloud-us/) illustrates an isolated environment designed for customers with shared government and compliance requirements. Whether a specific offering satisfies the NIST community-cloud definition depends on its access and governance model.

</details>

A **hybrid cloud** combines two or more distinct cloud infrastructures. Sensitive images could, for example, remain in a private cloud while additional processing capacity is temporarily obtained from a public cloud. This can combine control with elasticity, but it also requires compatible interfaces, secure data transfer and coordinated monitoring across the involved environments. [1, 6]

<details>
<summary><strong>Popular example: Cloud bursting (hybrid cloud)</strong></summary>

An application normally runs in a private cloud but temporarily uses resources from a public cloud when the local capacity is insufficient. The two clouds remain distinct, while data or application components have to move between them through compatible and secure interfaces [1].

#### 1.3.2 Is Every Remote Server a Cloud?

With the definitions introduced above, the question from Section 1.2 can now be answered. A server does not become a cloud simply because it is located elsewhere, can be reached through the network or runs as a virtual machine. These properties may be part of a cloud service, but they are not sufficient on their own. [1, 4]

The public Linux machine from Step 2 provides broad network access, but its resources remain fixed and have to be changed manually. The rented VPS from Step 3 transfers the physical hardware to a provider and may be easier to resize, but it can still consist of one manually provisioned server with fixed capacity. In both cases, resource pooling, rapid elasticity, on-demand self-service and measured usage may be missing. Conventional remote hosting and cloud computing can therefore look similar from the outside while being operated differently. [1, 6]

The transition in Step 4 becomes cloud computing when resources are offered as a service, provisioned on demand, drawn from a shared pool, scaled according to demand and measured. Other definitions may additionally emphasize virtualization, pay-per-use, service-level agreements or quality of service. The exact boundary therefore depends on the selected definition, but remote access alone is not enough. [1, 3, 4, 5]



### 1.4 Anatomy of a Cloud Application

Sections 1.2 and 1.3 described where the image-processing service can run and which parts are controlled by the provider. The next step is to follow one upload through the application itself. The service uses an asynchronous workflow: the API accepts the image and returns a job identifier before a worker has completed the processing. This keeps the public request short and allows the processing capacity to be changed independently. [7, 9, 20]

<p align="center"><img src="../../.admin/img/05_image_service_architecture.svg" alt="Architecture and request flow of the cloud image-processing service." width="78%"></p>

#### 1.4.1 Public Endpoint

The client starts with a domain such as `images.example.org`. DNS resolves the name to an IP address, after which the client establishes a TLS connection and sends an HTTP request. The public endpoint remains stable even when the internal application instances change. DNS, TLS and HTTP therefore solve different parts of the connection: finding the service, protecting the connection and defining the message exchanged with the service. [8, 9, 11]

Steps:

1. Resolve the service domain through DNS
2. Establish a TLS connection to the public endpoint
3. Send `POST /jobs` with the image and the requested operation

#### 1.4.2 Load Balancer

Several API instances can provide the same endpoint. The load balancer receives each request, selects one registered instance and forwards the request to it. Health checks allow an unavailable instance to be removed from the set of possible targets. The client does not need to know which instance was selected. A load balancer can therefore distribute traffic and prevent one failed API instance from stopping the service, but only while at least one healthy target remains. [7, 21]

Steps:

1. Receive the request at the public endpoint
2. Check which API instances are healthy
3. Select one healthy instance
4. Forward the request
5. Return the API response through the same public endpoint

#### 1.4.3 API

The API instance interprets the HTTP method, target, fields and message body. It rejects an invalid media type, an image that is too large or a request that is missing required parameters. Authentication and authorization are added in Section 1.5. For a valid request, the API creates a job identifier, stores the original image, records the initial job state and places a small job message in the queue. The response contains the identifier and a URI at which the current state can be requested. [9, 20]

Steps:

1. Validate the request
2. Create a job identifier
3. Store the image
4. Record the job as `queued`
5. Add the job identifier to the queue
6. Return `202 Accepted`, the identifier and the status URI

#### 1.4.4 Processing

The queue separates accepting a request from performing the expensive image operation. A worker retrieves one job, loads the stored image, creates the thumbnail and stores the result. It acknowledges the queue message only after the job has been completed. If the worker fails before the acknowledgement, the unfinished message can be returned to the queue. Several workers can consume from the same queue, which becomes the basis for the scalability experiment in Section 1.6. [7, 20]

Steps:

1. Retrieve one job from the queue
2. Mark the job as `processing`
3. Load the stored image
4. Generate and store the thumbnail
5. Mark the job as `completed` and record the result location
6. Acknowledge the queue message
7. On failure, mark the job as `failed` or return it to the queue

<p align="center"><img src="../../.admin/img/06_job_lifecycle.svg" alt="Lifecycle of an asynchronous image-processing job." width="70%"></p>

#### 1.4.5 Returning the Response

The first response does not contain the processed image. `202 Accepted` means that the request was accepted but has not yet been completed. The client keeps the job identifier and requests `GET /jobs/{job_id}` until the state changes from `queued` or `processing` to `completed` or `failed`. A completed job returns the location of the processed image. Accepting the upload and completing the image operation are therefore two separate events. [9]

Steps:

1. Receive `202 Accepted` and store the job identifier
2. Request the job state using the identifier
3. Continue waiting while the state is `queued` or `processing`
4. Retrieve the result when the state is `completed`
5. Handle the error when the state is `failed`

The application can now receive and process an image. The next section adds the controls required before this public endpoint can be used safely.


### 1.5 Securing the API: Identity and PKI

The public endpoint from Section 1.4 can receive requests from the Internet. Before the image is accepted, the service must answer three questions: who is making the request, whether that identity may use the operation and how the request can be protected while it travels through the network. [7, 16]

<p align="center"><img src="../../.admin/img/07_secure_api_request.svg" alt="Secure API request showing DNS, TLS, certificate validation, authentication and authorization." width="78%"></p>

An **identity** represents a user, device or application. **Authentication** verifies the claimed identity. **Authorization** then determines which actions that identity may perform. The two concepts are separate: a caller can be authenticated successfully but still not be allowed to upload or retrieve a particular image. [7, 12, 13, 16]

The communication itself requires **confidentiality** and **integrity**. Confidentiality prevents other parties on the network from reading the image and credentials. Integrity allows the receiver to detect whether the request or response was changed during transmission. TLS establishes this protected connection and also allows the client to authenticate the service it reached. [11]

The **public key infrastructure (PKI)** provides the required trust relationship. The service presents a certificate that connects its identity to a public key. The client checks the service name, the validity of the certificate and the path to an already trusted authority. The corresponding private key remains protected by the service. This allows the client to verify the service without knowing it in advance. [14, 15]

The complete request follows the sequence shown in the figure:

1. Locate the service through DNS
2. Establish a protected connection and verify the service identity
3. Present the client's identity
4. Authenticate the client
5. Authorize the requested operation
6. Execute the operation and return the protected response

None of these concepts replaces another. Encryption does not determine who may use the API, authentication does not grant every permission, and authorization does not protect the image while it is transmitted. Protected communication, authentication and authorization therefore have to work together at the public endpoint. [7, 11, 16]

Symmetric encryption is not affected in the same way. Post-quantum methods are therefore being standardized for future key establishment and digital signatures. [19, 22]

#### 1.6.3 Performance and Locality

A request spends time in the application and on the network. Greater distance usually increases the network round-trip time, while DNS resolution, the TLS handshake, queueing, processing and data transfer add further delay. DNS caching and TLS session resumption can reduce some repeated setup work, but they do not remove the physical network path. [6, 8, 11]

The simulation deliberately uses a smaller model. It assumes that one request occupies capacity for the selected processing time plus one measured round-trip time:

\[
\text{capacity} =
\frac{1000}{\text{processing time (ms)} + \text{round-trip time (ms)}}
\quad \text{requests/s}.
\]

The values in `region_latency_ms` can be replaced with median ping measurements to known server locations. Ping measures network round-trip time only; it does not measure DNS, TLS, queueing or application processing.

Simulation steps:

1. Select a small number of servers whose locations are known
2. Ping each server several times without producing excessive traffic
3. Calculate the median round-trip time for each location
4. Enter the values in `region_latency_ms`

#### 1.6.4 Content Delivery Network

The processed image can be copied to a CDN edge close to the client. The first request is a **cache miss** and still reaches the origin. A later request can become a **cache hit** and be returned directly from the edge while the stored response remains fresh. HTTP cache directives determine whether a response may be stored, how long it remains fresh and which request fields belong to its cache key. [9, 10]

The benefit therefore depends on the cache-hit rate and on the difference between the edge and origin response times. Newly uploaded images, personalized responses and content that is not safe to cache still reach the origin. A CDN can reduce the delivery work for repeated results, but it does not process a new image and does not create a second operational backend region. [10]

#### 1.6.5 Component Failure: Availability and Reliability

The elastic service can run several workers at the same time. These workers also provide redundancy: if one fails, the others can continue to consume jobs from the queue. An unacknowledged job can be returned to the queue instead of being treated as completed. However, the failure still reduces the total capacity. If the remaining workers process fewer requests than arrive, the backlog grows even though the service remains reachable. [7, 20]

This separates availability from performance. Redundancy can keep the service available during a component failure, but it does not guarantee that the original throughput is maintained. Monitoring is required to detect the failed worker and to restore the intended capacity. [7, 17]

#### 1.6.8 Choosing an Architecture

No architecture is best for every workload. Each mechanism introduced in the simulations removes one limitation and introduces another cost or responsibility.

| Requirement | Useful mechanism | New trade-off |
|---|---|---|
| Variable demand | Queue and elastic workers | Startup delay, scaling rules and peak cost |
| Repeated delivery to distant clients | CDN | Only cacheable content benefits; invalidation is required |
| Failure of one worker | Multiple workers and a durable queue | Additional capacity and monitoring |
| Failure of one region | Replication and regional failover | Higher cost, data consistency and recovery testing |
| Sensitive data | Suitable service model, deployment model and contractual controls | Less direct control or less elasticity, depending on the choice |

The decision therefore starts with the workload and the data rather than with a provider product. Demand variation determines the required capacity, client location influences latency, the business impact determines the RTO and RPO, and the data determines the acceptable provider trust and deployment model. Skills, migration effort and total operating cost then decide whether a local machine, a rented server or a cloud service is the appropriate solution. [1, 3, 6, 16, 17]
