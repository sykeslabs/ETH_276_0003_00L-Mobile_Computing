### 3.1 Repository 1: k3s-io/k3s-ansible

- **Repo:** https://github.com/k3s-io/k3s-ansible
- **Showcase:**
  - **General:** Use this [seven-node Raspberry Pi K3s walkthrough](https://www.youtube.com/watch?v=N4bfNefjBSw) as an example of the physical cluster.
  - **Asymmetric compute:** Adapt this [hybrid ARM and x86 cluster](https://picluster.ricsanfre.com/docs/architecture/) to different Pi generations or accelerators.
  - **Load balancing:** Adapt this [four-Pi K3s API showcase](https://dinofizzotti.com/blog/2020-05-09-raspberry-pi-cluster-part-2-todo-api-running-on-kubernetes-with-k3s/) to show which Pi answers.
  - **Failure:** Reproduce this [six-Pi node failure demonstration](https://www.youtube.com/watch?v=ZyTLMnzehyU&t=1610s) by disconnecting one worker.

Notes:

- Smallest and clearest starting point.
- Suitable as the main repository for the notebook.
- The example service and experiments still have to be added.
- A simple setup may keep one controller as a single point of failure.

### 3.2 Repository 2: techno-tim/k3s-ansible

- **Repo:** https://github.com/techno-tim/k3s-ansible
- **Showcase:**
  - **General:** Follow this [high-availability K3s walkthrough](https://technotim.com/posts/k3s-etcd-ansible/) to build several controllers and workers.
  - **Asymmetric compute:** Use this [hybrid ARM and x86 example](https://picluster.ricsanfre.com/docs/architecture/) as a model for unequal workers.
  - **Load balancing:** Use this [load-balancer walkthrough](https://technotim.com/posts/k3s-ha-install/) to distinguish application and control-plane endpoints.
  - **Failure:** Adapt this [high-availability K3s video](https://www.youtube.com/watch?v=UoOcLXfa8EU) by removing one controller or worker.

Notes:

- Strongest option for a high-availability showcase.
- Includes more infrastructure than the basic repository.
- Useful when controller failure should be part of the notebook.
- Less suitable for the simplest first experiment.

### 3.3 Repository 3: ricsanfre/pi-cluster

- **Repo:** https://github.com/ricsanfre/pi-cluster
- **Showcase:**
  - **General:** Use the [project walkthrough](https://picluster.ricsanfre.com/docs/home/) to explore the complete home cluster.
  - **Asymmetric compute:** Use the documented [hybrid ARM and x86 architecture](https://picluster.ricsanfre.com/docs/architecture/) to compare unequal workers.
  - **Load balancing:** Use the [Cilium load-balancer setup](https://picluster.ricsanfre.com/docs/cilium/) to expose services across the cluster.
  - **Failure:** Use the [observability setup](https://picluster.ricsanfre.com/docs/observability/) to follow node, application and storage failures.

Notes:

- Strongest source for heterogeneous hardware ideas.
- Provides useful monitoring and visualisation concepts.
- Covers considerably more than the planned notebook requires.
- Better as a source of ideas than as the initial implementation.
