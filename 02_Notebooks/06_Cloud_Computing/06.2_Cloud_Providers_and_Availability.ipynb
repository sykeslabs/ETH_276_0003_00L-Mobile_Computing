### 2.1 Introduction

The previous notebook introduced cloud computing and the division of control and responsibility between the customer and the provider. Selecting a service model, however, does not yet select a provider. The same type of service may have a different price, may not be available in the required location or may place different parts of the system behind the provider boundary. [1]

Therefore, a provider has to be compared using the requirements of the application. Some examples are: 


- **Costs**
  - Compute, storage, requests, egress and support can be charged separately.
  - A lower service price may require more engineering work from the customer.
- **Locality**
  - The selected region determines where the service is offered, but processing, backups and replicas have to be checked separately.
  - A provider may operate a Swiss region without offering every service or hardware type there.
- **Security and confidentiality**
  - The provider protects the underlying infrastructure, while identities, access and application data partly remain the customer's responsibility.
  - For managed AI services, it also matters whether requests are recorded or used for model training.
- **Availability**
  - Several zones or regions make redundancy possible, but the application has to be configured to use them.
  - A resource deployed once in one zone remains a single point of failure.
- **Control and trust**
  - IaaS provides control over the operating system and software stack, while PaaS places more of this stack under the provider's control.
  - Contracts and documentation have to cover the parts that cannot be inspected or operated by the customer.

A useful comparison therefore starts with the workload and its hard requirements. The following profiles provide the same calculator inputs for this comparison.

### 2.2 Pricing Profiles

Prices can only be compared when the calculator inputs are the same. The following seven profiles fix the amount of compute, storage, requests or egress.

The most important terms are defined as follows:

- **On-demand pricing:** Resources are paid for as they are used without a long-term commitment.
- **Egress:** Data transferred from the cloud provider to the Internet or another external location.
- **Cached data:** Data delivered from a cache close to the user instead of directly from the original server.
- **Origin:** The server or storage system containing the original data behind a cache or CDN.
- **Excluded:** A resource marked as excluded is not part of the stated price estimate.
- **vCPU:** A virtual processor assigned to a virtual machine or managed application instance.
- **Persistent system disk:** Storage containing the operating system and files that remain available after a virtual machine is stopped.
- **Prompt caching:** Previously processed model input is reused to reduce repeated processing and token costs.
- **Object storage:** Files are stored as objects in buckets and accessed through an API.
- **Hot, standard and cold storage:** Hot and standard storage provide immediate access for regularly used data, while cold storage costs less but may introduce retrieval costs or conditions.
- **Zone redundancy:** Data or services are replicated across several availability zones within one region.
- **GPU-hour:** The use of one GPU for one hour.
- **Retrieval:** Reading data from a cold or archive storage class.
- **Early deletion:** Removing cold or archive data before its minimum storage duration, which may cause an additional charge.
- **Egress excluded:** The estimate does not include the cost of transferring data out of the provider.
- **Operations excluded:** The estimate does not include charges for API operations such as uploading, downloading or listing objects.

AI models can provide different results, GPU types have different performance and storage classes have different retrieval conditions. Prices and regional availability can also change and have to be recalculated for a real deployment. [5, 9, 13, 15]
