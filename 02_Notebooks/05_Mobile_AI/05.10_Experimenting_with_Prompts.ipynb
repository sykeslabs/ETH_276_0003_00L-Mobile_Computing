### 1. Running a Model

We assume that we are logged in to the Pi with SSH and password "password":

```bash
ssh cas@172.31.250.1
```

You can run models using the standard **Ollama CLI commands** (CLI=Command Line Interface) on the terminal. However, to make experimenting with parameters easier, we provide a custom command called `ollama-gen`. This command acts as a lightweight wrapper around Ollama's REST API `generate` endpoint, allowing you to adjust parameters directly from the command line.

```sh
ollama-gen --model <MODEL> --prompt "<PROMPT>" [optional parameters]
```

To see all available options / optional parameters, and to learn more about the custom command, run `ollama-gen --help`.

All flags and options can appear in any order. When experimenting, it is often useful to set `num_predict` to limit the number of generated tokens. Since this command is purely CLI-based, you can safely interrupt the model at any time with `CTRL + C`.

Unlike the raw Ollama API response, the `ollama-gen` command returns a **streamed, formatted, and simplified JSON object**. All durations are converted to ms, and throughput rates are computed automatically. For readability, internal fields such as the context vector are removed. The first line echoes the **exact request parameters** so you can confirm that your settings took effect. The output format closely mirrors the behavior of the CLI command `ollama run <MODEL> <PROMPT> --verbose`.

### 2. Quality and Runtime with Blank, Simple, and Advanced Prompting

As a first step, we evaluate both the **output quality** and the **runtime performance** of different models. While quality is subjective, runtime and throughput (`token/s`) provide measurable insight into model efficiency. For each of the following tasks, consider the following reported metrics:
- total duration (ms)
- load duration (ms)
- prompt eval duration (ms)
- prompt eval rate (tokens/s)
- eval duration (ms)
- eval rate (tokens/s)

#### 2.1. Blank Prompting

We start with a blank prompt, giving the model an empty input to see how it behaves without instructions. This reveals its **default tone, formatting, and training biases**.

<div class="alert alert-block alert-info">
Run each command five times and compute the average values of the above metrics. Note that we have to use the custom command here, as model rambling can be quite severe without any guardrails.
</div>

```sh
ollama-gen --model gemma3:270m --prompt " " --num_predict 50

ollama-gen --model qwen3:1.7b --prompt " " --num_predict 50

ollama-gen --model qwen3:1.7b --prompt " " --think false --num_predict 50

ollama-gen --model tinyStories:110m --prompt " " --num_predict 50
```

If you see `done_reason: "length"`, the model hit the `num_predict` cap. It was cut off by the hard limit, not by a `stop` sequence.

#### 2.2. Simple Prompting
Next we use a simple, factual question to see how well each model follows instructions and delivers concise answers. At this stage, we can move away from the API and instead use the Ollama CLI directly. Once again, we compare both **response quality and runtime efficiency**.

<div class="alert alert-block alert-info">
Run each of the following commands once and note the values of the above metrics. While this is not statistically robust, it gives us a rough indication of each model's behavior. If you want more reliable values, rerun the same prompt multiple times and average the collected data. To start, choose a simple and restrictive question that can be answered briefly.
</div>

**Example Prompt**: "What is the capital of Switzerland? How was this decided? Keep it to a maximum of 5 sentences."

```sh
ollama run gemma3:1b-it-qat "What is the capital of Switzerland? How was this decided? Keep it to a maximum of 5 sentences." --verbose

ollama run qwen3:1.7b "What is the capital of Switzerland? How was this decided? Keep it to a maximum of 5 sentences." --verbose

ollama run tinyStories:110m "What is the capital of Switzerland? How was this decided? Keep it to a maximum of 5 sentences." --verbose
```

*Hint*: If TinyStories tends to ramble, you can either let it finish naturally (this may take up to about 4 minutes if you are curious) or use the custom command to limit the output tokens to 2048.
