### 1.1 Introduction

Data is everywhere: in music, video, temperature readings, heart-rate measurements, GPS positions, and text messages. To store, process, or transmit information, we first need a way to represent it. Two important types of representation are:

- Analog: values are represented on a continuous scale. Between any two distinct values, there are always other possible values.
- Digital: values are represented using discrete symbols or steps, often as sequences of `0`s and `1`s.

The difference is not simply whether something changes over time. An analog value can be constant, such as a steady voltage or an unchanging temperature. Likewise, digital data can describe a time-varying signal, as an audio file does. In communication systems, the chosen representation affects how information is transmitted, how noise affects it, and how easily it can be stored, copied, or processed. This notebook builds intuition for analog and digital representations using illustrated examples.

### 1.2 Analog Signals

A signal is a representation that carries information. It may be a physical phenomenon, such as a voltage, a radio wave, light, or sound pressure, but it can also be a mathematical model of such a phenomenon.

An analog signal is continuous-valued: its amplitude, phase, frequency, or another measured quantity can take values on a continuous scale. It does not have to change over time. A constant voltage, for example, is still analog if its value is represented continuously.

- Analog signals are continuous-valued, not necessarily time-varying.
- When they do vary over time, their values can change smoothly rather than jumping between fixed levels.
- Everyday examples include your voice when you speak, the voltage in an electrical circuit, the brightness of a light source, or the temperature in a room.

Analog signals are important in transmission because many physical media naturally carry continuous quantities. Even when information starts as digital data, it is often converted into an analog waveform before it travels through a wire, antenna, optical fiber, or speaker.

### 1.3 Digital Data

Data is information represented in a form that can be stored, analyzed, interpreted, or transmitted. Digital data uses discrete symbols or values. In computers and digital communication systems, these symbols are usually represented as sequences of `0`s and `1`s.

Digital data can represent many kinds of information: numbers, text, images, sensor readings, or sound recordings. Because the representation is discrete, digital data can often be stored, copied, compressed, and protected against noise with high reliability and no loss.



Digital data is distinct from the physical signal that carries it. For example, an MP3 file contains digital data. When the file is played through a speaker, the speaker produces an audible analog sound-pressure wave.

To label the positions in a digital sequence, we use the natural numbers, including zero:
$$
\mathbb{N}_0 = \{0,1,2,3,\dots\}.
$$
The subscript 0 makes it explicit that zero belongs to this set. A digital bitstream can then be described as follows:
$$
s[n] \;\in\;\{0,1\}, 
\quad n \in \mathbb{N}_0
$$
A digital sequence can also use more than two possible symbols. For example, a sequence of 8-bit symbols can be described as follows:
$$
s[n] \;\in\;\{0,1,2,\dots,255\}, 
\quad n \in \mathbb{N}_0
$$
Here:

- $n$ is the discrete symbol index.
- $s[n]$ is the value of the symbol at index $n$.
- A binary symbol has two possible values, while an 8-bit symbol has 256 possible values.



If the digital data forms a time series, the index can be linked to physical time. Let $T$ be the sampling or symbol interval in seconds. The value at index $n$ then occurs at time
$$
t_n = nT, 
\quad n \in \mathbb{N}_0
$$
For a binary time series, the possible values are
$$
s[nT] \;\in\;\{0,1\}, 
\quad n \in \mathbb{N}_0
$$
and for 8-bit symbols, they are
$$
s[nT] \;\in\;\{0,1,2,\dots,255\}, 
\quad n \in \mathbb{N}_0
$$
In this form, the horizontal axis shows time in seconds, while the possible values of $s[nT]$ remain discrete. In the examples below, we set $T = 1\,\mathrm{s}$, so the samples occur at $0\,\mathrm{s}$, $1\,\mathrm{s}$, $2\,\mathrm{s}$, and so on.

By mapping each symbol $s[n]$, or each time-series value $s[nT]$, onto a pulse or level in a continuous-time waveform, we turn the digital sequence into a physical signal suitable for transmission or further processing. 


### 1.4 The Roles of Analog and Digital Representations

Analog and digital representations serve different roles in the same system. We observe the physical world through quantities such as voltage, pressure, acceleration, temperature, light intensity, and radio field strength. These quantities are usually observed as analog signals: they can take many possible values, and their measured values depend on the sensor, the medium, noise, and the measurement circuit.

Digital systems are useful when we want to store, process, copy, compress, encrypt, compare, or transmit information reliably. A phone, for example, may start with an analog voltage from a microphone, convert it into samples, compress those samples into a digital audio stream, protect the stream with error-correcting codes, and then transmit it using an analog electromagnetic wave. At the receiver, the process is reversed, and a speaker produces an analog sound-pressure wave again.



> The word **digital** comes from the Latin word *digitus*, meaning a finger or toe. Because people have long counted on their fingers, **digit** came to mean a numeral. Digital data represents information using distinct symbols or digits, commonly 0 and 1.
>
> **Discrete** means separate and distinct rather than continuous. A discrete set contains individually distinguishable possibilities, such as the six outcomes of a die. By contrast, a continuous quantity can take any value within a range. Digital data is discrete, although discrete information does not have to be binary.
