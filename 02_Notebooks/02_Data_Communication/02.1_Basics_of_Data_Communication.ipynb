### 1.1 Introduction

Communication systems are everywhere and have changed daily life. Nearly everyone has listened to the radio, used a USB cable or a Bluetooth headset, and owns a smartphone. Storing files on a USB stick or in the cloud is just as routine. All of these are instances of the same thing: a communication system.

### 1.2 The Chain from Source to Sink

A satellite link, a Bluetooth headset, and a USB stick look nothing alike, yet they perform the same sequence of operations on the information they carry. Virtually all communication systems contain the same main building blocks. Only the implementation details within these blocks differ from system to system.

<div class="alert alert-light">

**Definition.** A communication system sends information from a transmitter to a receiver over space and/or time.

</div>

The diagram below shows these building blocks. The blocks on the sending side form the transmitter (Tx), the blocks on the receiving side form the receiver (Rx), and the channel sits between them. Every block in the transmitter has a corresponding block in the receiver that reverses its work. Several of the blocks below deal with redundancy, meaning the parts of a message that repeat or are predictable, and that therefore add length without adding information.

<p style="text-align: center;"><img src="../../.admin/img/communication_system_overview.png" alt="Overview of a communication system and its main building blocks." style="max-width: 60%; height: auto;"></p>

**Source:** generates the information to be sent, such as a voice, a photo, a temperature reading, or a bank balance. The information may be analog, such as speech, or already digital, such as a file. The source may be a person, a program, or a sensor.


**Source Encoder:** represents that information as compactly as possible, using as few bits as it can, by removing redundancy that is already present in the data.


**Channel Encoder:** adds a different, controlled redundancy back, in a form the receiver knows in advance, so that errors introduced later can be detected and, where possible, corrected.

**Modulator:** the interface between the transmitter and the channel. It converts the encoded data into a physical signal the channel can carry, such as a radio wave, a pulse of light, or a sound, using the frequencies, time slots, and power level assigned to the system.

**Channel:** the physical medium that carries the signal over space and/or time. Cables, optical fiber, air, and vacuum carry information over space. Storage media such as a USB stick, a hard drive, or a printed QR code carry it over time, because the information is written now and read later. Whatever the medium, the signal is degraded on the way. A radio signal weakens with distance, picks up noise from the electronics at both ends, and is disturbed by other systems using the same frequencies at the same time. A storage medium ages, gets scratched, or is smudged.

**Demodulator:** the interface between the channel and the receiver. It picks up the physical signal and reconstructs the data from it. Because the signal arrives degraded, the demodulator has to decide which bits were most likely sent, and it sometimes decides wrong.

**Channel Decoder:** uses the redundancy added by the channel encoder to detect those wrong decisions and, where possible, correct them.

**Source Decoder:** reverses the source encoder. Some source encoders discard detail that the sink is unlikely to notice, which is why the diagram above labels the output estimated information rather than the original information.

**Sink:** receives the information. Like the source, it may be a person, a program, or a device.

The source encoder and the channel encoder treat redundancy in opposite ways, for opposite reasons. The source encoder removes redundancy that is already in the data, to make it smaller. A photo of a clear sky, for example, contains thousands of near-identical neighboring pixels, which can be replaced by a short instruction to repeat that color. The channel encoder then adds a different redundancy back, such as a short summary of the message that lets the receiver check whether what arrived still matches what was sent. Stronger versions let the receiver repair some of the damage without asking for anything to be sent again.

The figure above shows how a single message is prepared, sent, and recovered. This is only part of a real communication system. Most systems also organize their functions into layers. The diagram below shows a simplified five-layer version of the Open Systems Interconnection (OSI) model: application, transport, network, data link, and physical.

The middle of the chain, from the channel encoder through to the channel decoder, is the physical layer, marked in the diagram below as the physical channel. The source encoder and source decoder usually sit higher up, close to the application that produced the information. Everything above the physical layer, such as addressing, routing, and splitting a message into packets, belongs to the layers above and is the subject of a computer networks course. This course begins at the physical layer.

### 1.4 Design Goals of a Communication System

There is no single communication system suited to every application, because different applications demand different things. Common design goals are as follows:

| Design goal | What it means | Typically measured in |
|---|---|---|
| Throughput | How much information is moved per second, also called the data rate | megabits per second |
| Latency | How long information takes to travel from the source to the sink | milliseconds |
| Reliability | How many errors occur, and how well the system tolerates them | errors per transmitted bit |
| Range and coverage | How far apart transmitter and receiver can be, and how large an area is served | meters or kilometers, square kilometers |
| Power | How much energy the transmitter and receiver consume | watts |
| Size and weight | How large and how heavy the equipment has to be | cubic centimeters, grams |
| Spectral efficiency | How much information fits into a given range of frequencies | bits per second per hertz |
| Cost | Hardware, deployment, and the license fees for using a frequency band | francs |
| Mobility | Whether the system keeps working while the transmitter or receiver is moving, and how fast | kilometers per hour |
| Security | How difficult it is for someone else to intercept or disrupt the transmission | time needed to break the protection |

Designers normally target several of these at once, which makes the design of a communication system a compromise rather than the optimization of any single goal. No system scores well on all of them.

### 1.5 Same Blocks, Different Trade-offs

Because these goals conflict, each technology maximizes a few of them and accepts the cost on the rest. The building blocks stay the same; what changes is how each block is built to serve those priorities.

The clearest example is throughput against size, weight, power, and cost. A system built for maximum throughput, such as Wi-Fi, needs complex signal processing. That processing consumes power, which means a larger battery and a more expensive device. A system built for small size, low weight, long battery life, and low cost, such as a battery-powered sensor, cannot afford that processing. It also uses a much narrower slice of spectrum and far less transmit power, which is the deeper reason it cannot reach the same throughput. Neither system is badly designed. They were built to different goals.

| Technology | Maximizes | Accepts in return |
|---|---|---|
| GNSS, such as GPS | Worldwide coverage | Very low throughput, one-way only |
| Bluetooth Low Energy | Power efficiency, low cost | Short range, small amounts of data |
| Wi-Fi | Throughput | Short range, higher power, shared spectrum |
| Cellular | Coverage, mobility | Cost of licensed spectrum and infrastructure |

A GPS receiver shows how far this can go. It only listens, and can never tell the satellite that something arrived damaged, so asking for a retransmission is not an option. Everything needed to recover the signal has to be built into the transmission in advance. In exchange for working almost anywhere on the planet with a clear view of the sky, it delivers roughly fifty bits per second, a tiny fraction of what a Bluetooth headset carries across one meter.

The useful question is therefore never which technology is best, but which sacrifices are acceptable in a given situation. Recognizing which goals a technology gave up, and why, is the skill the rest of this course develops.
